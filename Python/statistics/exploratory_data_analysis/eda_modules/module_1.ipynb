### Module 1.1: Algorithmic Thinking

At its heart, all code is just a set of instructions. An **algorithm** is simply the formal name for "a finite sequence of well-defined, computer-implementable instructions... to solve a class of problems."

This is a fancy way of saying "a recipe".

Your first job as a Statistician-Pythonista is to be a *chef*, not a *line cook*. A line cook just follows the recipe. A chef *designs* the recipe.

To design a recipe, you first write it in a human-readable format, not a machine-readable one. This is **pseudocode**.

* **Pseudocode (The Idea):**
    `for each number in my list:`
    `if the number is greater than 10:`
    `add it to a new list of 'large numbers'`

* **Code (The Implementation):**
	```python
    large_numbers = [num for num in my_list if num > 10]
	```

You must *always* solve the problem in pseudocode (in your head or on paper) *before* you write a line of Python. This separates the *logic* of your analysis from the *syntax* of the language.

Once that skeleton exists, flesh it out with a few deliberate checks:

* **State the goal in plain language:** "I want to keep only the outliers whose Z-score is greater than 3." If you cannot say it clearly, you will not code it cleanly.
* **Work a toy example by hand:** Take 5 or 6 friendly numbers, run through your recipe, and make sure the output matches what you expect. This guards against hidden assumptions.
* **List the data requirements:** Does your logic assume sorted data, numeric types, or non-missing values? Capture these in comments or assertions so future-you remembers the preconditions.

Only after those steps should you start translating into Python syntax. If you change your mind halfway, adjust the pseudocode first and let the implementation follow. Great statisticians treat code like a mathematical proof: start with a logical outline, verify it on a miniature case, then generalize with confidence.

### Module 1.2: Algorithmic Complexity (Big O Notation)

This is the most critical concept for moving from an amateur to a professional.

Algorithmic complexity, or **Big O Notation**, does *not* measure how fast your code runs in seconds. It measures how your code's runtime *scales* as your input data gets larger.

Let $n$ be the size of your input (e.g., the number of rows in your DataFrame).

#### $O(1)$ â€” Constant Time
This is the holy grail. The operation takes the same amount of time regardless of the size of your data.
* **What it is:** A single, direct lookup.
* **The Math:** The time $T$ is a constant $c$: $T(n) = c$.
* **The Code:** Accessing an item in a list by its index (`my_list[100]`) or a dictionary by its key (`my_dict['user_id']`). It takes the same time to get `my_list[100]` whether the list has 200 items or 200 million items.
* **Why it works (for dicts):** A dictionary uses a **hash function**. It converts the key (e.g., `'user_id'`) into a memory address (an index). It does not scan the dictionary; it *jumps* directly to the correct location.


#### $O(n)$ â€” Linear Time
This is the most common and acceptable "cost" for many operations. The runtime grows in a 1-to-1 relationship with the size of your input.
* **What it is:** You must "touch" every item in your list once.
* **The Math:** The time $T$ is proportional to $n$: $T(n) = c \times n$. If you double $n$, you double the time.
* **The Code:** A single `for` loop. Searching for an item in an unsorted list (e.g., `if 'Jhonatan' in my_list:`). Calculating the mean of a `numpy` array (it must add *every* number).


#### $O(n^2)$ â€” Quadratic Time
This is the "danger zone" and a primary source of slow code. The runtime grows exponentially.
* **What it is:** You must compare every item in your list to *every other item* in your list.
* **The Math:** The time $T$ is proportional to $n$ squared: $T(n) = c \times n^2$. If your data grows by 10x, your runtime grows by **100x**.
* **The Code:** A nested `for` loop.
    ```python
	for item_a in my_list:
    for item_b in my_list:
    if item_a == item_b:
    ...
	```

#### $O(\log n)$ â€” Logarithmic Time
This is the gold standard for searching. It is incredibly fast. The runtime grows *slower* as your data gets bigger.
* **What it is:** You cut the search area in half with each step.
* **The Math:** The time $T$ is proportional to the logarithm of $n$: $T(n) = c \times \log(n)$. If your data doubles from 1 million to 2 million, your time increases by *one single step*.
* **The Code:** A **binary search** on *sorted* data. This is how database indexes work.


**How this informs your "Statistician-Pythonista" decisions:**

* **Why we use a `dict` (or `set`) for lookups:** When you ask "is this `user_id` in my list of 10 million banned users?", a `list` ($O(n)$) would take minutes. A `set` ($O(1)$) takes microseconds. You have just changed the complexity from linear to constant.
* **Why we use `numpy`:** `NumPy` operations are "vectorized". A `for` loop in Python ($O(n)$) is slow because the *interpreter* is slow. A `numpy` operation (`my_array * 2`) is also $O(n)$, but the loop is run in pre-compiled C code, making the constant $c$ *thousands* of times smaller.

When evaluating complexity, think about three complementary lenses:

1. **Worst-, average-, and best-case behavior:** Sorting a list with `sorted()` is $O(n \log n)$ in the worst case, but if the list is almost sorted the constant factors shrink dramatically. Know which scenario matches your data before hand-optimizing.
2. **Practical breakpoints:** Back-of-the-envelope calculations help decide when to refactor. If a quadratic routine takes 0.1 seconds on 1,000 rows, it will take roughly 10 seconds on 10,000 rows and **over 15 minutes** on 100,000 rows-that is one coffee break per run.
3. **Space as well as time:** Storing intermediate copies of large DataFrames doubles your memory footprint. Prefer in-place transforms (when safe) or generators when memory is the scarce resource.

A quick heuristic table you can keep in mind:

| Data size ($n$) | $O(1)$ | $O(\log n)$ | $O(n)$ | $O(n \log n)$ | $O(n^2)$ |
|-----------------|--------|-------------|--------|---------------|----------|
| $10^3$          | instant | instant    | fast | fast | acceptable |
| $10^5$          | instant | instant    | seconds | seconds-minutes | painful |
| $10^7$          | instant | milliseconds | minutes | impractical | infeasible |

Use the table not as gospel but as intuition. Before you run a heavy analysis, sketch its complexity and estimate the cost-especially when collaborating on shared compute resources. Your teammates (and future self) will thank you.

### Module 1.3: Readability & The "Zen of Python"

Code is read far more often than it is written. Your primary audience is not the computer; it is the next human to read your code (which is often you, six months from now, when you have forgotten everything).

Python has a guiding philosophy for this. You can read it by typing `import this` into a Python prompt.

> **The Zen of Python, by Tim Peters**
>
> Beautiful is better than ugly.
> Explicit is better than implicit.
> Simple is better than complex.
> Complex is better than complicated.
> Flat is better than nested.
> Sparse is better than dense.
> **Readability counts.**
> ...
> If the implementation is hard to explain, it's a bad idea.
> If the implementation is easy to explain, it may be a good idea.

This is not just poetry. It is a set of design principles.
* "Explicit is better than implicit": Name your variables clearly. `rooms_per_household` is better than `rph`.
* "Flat is better than nested": Avoid deeply nested `if` statements and loops. A 4-level nested loop is a sign of $O(n^4)$ complexity and is unreadable.
* "Readability counts": This is the law.

Actionable extensions of those aphorisms:

* **Tell a story with names:** Pick verbs for functions (`clean_missing_values`) and nouns for data (`customer_orders_df`). Consistent naming lowers the cognitive load when you revisit a notebook after weeks away.
* **Let whitespace breathe:** Group related operations with blank lines and align arguments in multi-line function calls. Visual structure steers readers through your reasoning.
* **Document intent, not mechanics:** A short comment like `# Convert to log-scale to stabilize variance` explains *why* a transform exists. Avoid narrating obvious code such as `# increment i`.
* **Prefer guard clauses over deep nesting:** Instead of indenting three levels, bail out early with `if df.empty: return df`. Readers see the "happy path" without spelunking through conditions.
* **Lint once, read forever:** Tools like `black`, `ruff`, or `flake8` automate consistency so you can invest brainpower in analysis rather than spacing.

Before moving on, read your notebook aloud (or use `nbconvert` to HTML and skim it). If the prose feels choppy or the code blocks look dense, polish them. Clear exposition turns exploratory analysis into reusable knowledge.

### Module 1.4: Abstraction & The DRY Principle

**DRY** stands for **"Don't Repeat Yourself"**.

This is the core principle of software hygiene. If you find yourself copying and pasting a block of code, you have created a *maintenance nightmare*.

* **The Problem (WET Code: "Write Everything Twice"):**
    ```python
	# Block 1
    raw_data = data['col_a']
    mean = raw_data.mean()
    std = raw_data.std()
    z_scores = (raw_data - mean) / std

    # ... 100 lines later ...

    # Block 2
    raw_data_2 = data['col_b']
    mean = raw_data_2.mean()
    std = raw_data_2.std()
    z_scores_2 = (raw_data_2 - mean) / std
	```

* **The Issue:** You just found a bug. You should have used $n-1$ (sample standard deviation). Now you have to find and fix this logic in *two* (or ten) different places.

* **The Solution (DRY Code: Abstraction):**
    You write *one* function that holds the "truth" of your algorithm.
    def calculate_z_scores(series):
    ```python
	"""Calculates the standard Z-scores for a pandas Series."""
    mean = series.mean()
    std = series.std(ddof=1) # The correct sample std. dev.
    return (series - mean) / std

    # Block 1
    z_scores = calculate_z_scores(data['col_a'])

    # Block 2
    z_scores_2 = calculate_z_scores(data['col_b'])
	```
Now, when you find a bug (like the `ddof=1`), you fix it *once*, inside the function. Every part of your code that uses this function is instantly corrected. This is the essence of abstraction.

To practice DRY thinking:

1. **Scan for repetition deliberately:** When you copy-paste code, pause and ask whether a helper function, a loop, or a dictionary could express the pattern once.
2. **Parameterize the differences:** If two blocks only vary by column name or a threshold, turn those parts into parameters with sensible defaults.
3. **Keep abstractions slim:** A reusable function should do one focused task. If it requires six flags to toggle behavior, split it into specialized helpers instead.

DRY code is easier to test, easier to reason about, and easier to share. In data science teams, it also means experiments become reproducible pipelines rather than fragile one-off notebooks.


### Module 1.5: Functions — The Building Blocks of Abstraction

In

**1. Anatomy of a Good Function:**

A well-written function has several key components:

```python
import typing

def calculate_z_scores(
    series: pd.Series, # Input data: expects a pandas Series
    ddof: int = 1      # Delta degrees of freedom for std dev (1 for sample)
) -> pd.Series:        # Return type: indicates it gives back a pandas Series
    """Calculates the standard Z-scores for a pandas Series.

    Z-score is defined as: z = (x - mean) / std_dev

    Args:
        series: The input pandas Series containing numerical data.
        ddof: Delta Degrees of Freedom for the standard deviation calculation.
              Defaults to 1 for sample standard deviation.

    Returns:
        A pandas Series containing the Z-scores.

    Raises:
        TypeError: If the input is not a pandas Series.
        ValueError: If the series standard deviation is zero.
    """
    # Input Validation
    if not isinstance(series, pd.Series):
        raise TypeError("Input must be a pandas Series.")

    # Core Logic
    mean_val = series.mean()
    std_val = series.std(ddof=ddof)

    if std_val == 0:
        # Avoid division by zero; return NaNs or zeros as appropriate
        # Here, returning zeros might be misleading, NaNs indicate undefined score
        return pd.Series(np.nan, index=series.index)

    z_scores = (series - mean_val) / std_val

    return z_scores

```

**Breakdown:**

  * **`def calculate_z_scores(...)`:** The function definition keyword (`def`), a descriptive name (verb phrase), and parentheses for parameters.
  * **Parameters (`series: pd.Series`, `ddof: int = 1`):**
      * **Type Hints (`:`):** `series: pd.Series` tells humans and tools (like type checkers) that this function *expects* a pandas Series. `ddof: int` expects an integer. This catches errors *before* you run the code.
      * **Default Arguments (`=`):** `ddof: int = 1` makes `ddof` optional. If the caller doesn't provide it, it defaults to 1 (calculating the *sample* standard deviation, which is usually correct in statistics).
  * **Return Type Hint (`-> pd.Series`):** This clarifies what the function *gives back*.
  * **Docstring (`"""..."""`):** Essential documentation. It explains:
    1.  What the function *does* (summary line).
    2.  The *math* or logic (optional but good here).
    3.  What each `Args` (parameter) is.
    4.  What it `Returns`.
    5.  What `Raises` (errors it might throw). Good tools can read this to provide help.
  * **Input Validation:** The `if not isinstance(...)` block is crucial. It prevents your function from crashing with weird errors later if someone passes in a list or a number instead of a Series. Fail early, fail explicitly.
  * **Core Logic:** The actual calculation. Note the use of `.mean()` and `.std(ddof=ddof)` from pandas.
  * **Handling Edge Cases:** The `if std_val == 0:` block prevents a `ZeroDivisionError`. Returning `NaN` (Not a Number) is statistically appropriate here â€“ Z-scores are undefined if there's no spread.
  * **Return Statement:** `return z_scores` passes the result back to the caller.

**2. Arguments vs. Keyword Arguments:**

  * **Positional Arguments:** `calculate_z_scores(my_data_series, 0)` relies on the *order*. The first value goes to `series`, the second to `ddof`. This is fragile if you add more parameters later.
  * **Keyword Arguments:** `calculate_z_scores(series=my_data_series, ddof=0)` uses the *names*. This is much more readable and robust. You can even change the order: `calculate_z_scores(ddof=0, series=my_data_series)`.

**Best Practice:** Use positional arguments only for the essential, primary inputs (like the data itself). Use keyword arguments for all optional parameters or configuration settings.

Beyond signatures, consider the *behavioral* qualities of great functions:

* **Keep them pure when possible:** A pure function depends only on its inputs and produces no side effects. That makes unit testing trivial and enables memoization or parallel execution.
* **Return early on invalid states:** Guard clauses such as `if series.empty: return series` make control flow obvious and prevent subtle bugs later on.
* **Surface assumptions:** If you expect strictly positive values or a specific unit of measure, either validate it or encode it in the function name (e.g., `calculate_log_z_scores`). Hidden assumptions become time bombs.
* **Pair functions with tests:** Even a quick `pytest` case that feeds in a miniature Series with known results will anchor the behavior and protect against regressions when you refactor.
* **Document side effects explicitly:** If a helper writes to disk, logs metrics, or mutates a global variable, say so in the docstring. Callers should never have to discover surprises by reading the implementation.

Thinking in functions is the foundation for building analysis scripts that scale into production-grade pipelines. Each well-designed function becomes a reliable building block in your analytical toolkit.

### Module 1.6: Core Data Structures in Action

Let's revisit our Big O discussion from Moment 1 and see how it dictates our choice of data structures in practice.

  * **Lists (`[]`): Ordered, Mutable, Allows Duplicates**

      * **Use Case:** Storing a sequence where order matters and items might change or repeat.
      * **Big O Link:**
          * `my_list[i]`: Access by index is $O(1)$ (fast).
          * `my_list.append(x)`: Adding to the end is *usually* $O(1)$, but occasionally $O(n)$ if the list needs to resize its internal memory.
          * `my_list.insert(i, x)`: Inserting in the middle is $O(n)$ (slow, everything after `i` must shift).
          * `x in my_list`: Checking for membership is $O(n)$ (slow, Python scans the list).
      * **Statistician-Pythonista Take:** Great for collecting results in order. Terrible for checking if you have already seen a specific item (like a `user_id`).

  * **Tuples (`()`): Ordered, Immutable, Allows Duplicates**

      * **Use Case:** Storing a fixed sequence that should *not* change, like coordinates `(x, y)` or RGB values `(255, 0, 0)`. Often used as dictionary keys because they are immutable.
      * **Big O Link:** Same as lists for access ($O(1)$), but you cannot append or insert. Membership check (`x in my_tuple`) is still $O(n)$.
      * **Statistician-Pythonista Take:** Use them when you want to signal "this group of values belongs together and won't change". Good for returning multiple fixed values from a function: `return mean, std_dev`.

  * **Dictionaries (`{}`): Unordered (in older Python), Mutable, Keys Must Be Unique & Immutable**

      * **Use Case:** Storing key-value pairs. The *fastest* way to look up information based on a unique identifier.
      * **Big O Link:**
          * `my_dict[key]`: Access by key is $O(1)$ on average (lightning fast, thanks to hashing).
          * `key in my_dict`: Checking for key membership is $O(1)$ on average.
          * Adding/updating a key-value pair is $O(1)$ on average.
      * **Statistician-Pythonista Take:** Your go-to for frequency counts, mapping IDs to names, storing configuration settings, or anytime you need rapid lookups based on an identifier.

  * **Sets (`{}` or `set()`): Unordered, Mutable, No Duplicates**

      * **Use Case:** Storing unique items when order doesn't matter. Primarily used for *very fast* membership testing and set operations (union, intersection).
      * **Big O Link:**
          * `x in my_set`: Checking for membership is $O(1)$ on average (just as fast as a dictionary key lookup).
          * Adding an item is $O(1)$ on average.
          * Set operations like union (`|`) and intersection (`&`) are typically very efficient, often better than linear time depending on sizes.
      * **Statistician-Pythonista Take:** If your only question is "Have I seen this `user_id` before?", use a `set`, not a `list`. It's the difference between waiting minutes ($O(n)$) and milliseconds ($O(1)$) on large datasets.

**Example: Efficiently finding unique items**

  * **Bad (List approach, $O(n^2)$ complexity due to repeated $O(n)$ checks):**
    ```python
    unique_items = []
    for item in large_list:
        if item not in unique_items: # This 'in' check is O(n) each time!
            unique_items.append(item)
    ```
  * **Good (Set approach, $O(n)$ complexity overall):**
    ```python
    unique_items = set(large_list) # Conversion to set is O(n)
    # Membership checks later using 'in unique_items' will be O(1)
    ```

Python's standard library gives you even more specialized tools:

* **`collections.defaultdict` and `Counter`:** Perfect for tallying categories without manual `if key in dict` checks. They keep code short and communicate intent immediately.
* **`deque`:** A double-ended queue with $O(1)$ appends and pops on both sides - ideal for rolling-window calculations when you need to drop old observations efficiently.
* **`array` or `numpy.ndarray`:** When memory pressure matters, these compact numeric containers avoid the overhead of Python objects and keep cache lines hot.

In pandas, remember that a `Series` is essentially an indexed dictionary with vectorized math. Choose the simplest native Python structure that solves the problem first, and move to pandas or NumPy only when you need labeled axes or heavy arithmetic. Data structure fluency lets you express analysis ideas in the most performant and maintainable way.

### Module 1.7: Pythonic Idioms — Writing Elegant & Efficient Code

"Pythonic" means leveraging the language's features to write code that is concise, readable, and often faster.

  * **List Comprehensions:** The concise way to create lists.

      * **Verbose:**
        ```python
        squares = []
        for x in range(10):
            squares.append(x * x)
        ```
      * **Pythonic:**
        ```python
        squares = [x * x for x in range(10)]
        ```
      * **With Condition:**
        ```python
        even_squares = [x * x for x in range(10) if x % 2 == 0]
        ```

  * **Dictionary Comprehensions:** Similar, but for creating dictionaries.

    ```python
    square_dict = {x: x * x for x in range(5)}
    # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
    ```

  * **`enumerate()`:** Get both the index and the value when looping.

      * **Verbose:**
        ```python
        index = 0
        for value in my_list:
            print(f"Index: {index}, Value: {value}")
            index += 1
        ```
      * **Pythonic:**
        ```python
        for index, value in enumerate(my_list):
            print(f"Index: {index}, Value: {value}")
        ```

  * **`zip()`:** Loop over multiple iterables simultaneously.

    ```python
    names = ["Alice", "Bob", "Charlie"]
    scores = [85, 92, 78]
    for name, score in zip(names, scores):
        print(f"{name}: {score}")
    # Stops at the end of the shortest iterable
    ```

  * **Unpacking (`*` and `**`):**

      * **Sequence Unpacking:**
        ```python
        first, second, *rest = [10, 20, 30, 40, 50]
        # first = 10, second = 20, rest = [30, 40, 50]
        ```
      * **Dictionary Unpacking (in function calls):**
        ```python
        def describe_user(user_id, name, age):
            print(f"User {user_id}: {name}, Age: {age}")

        user_info = {'user_id': 123, 'name': 'Jhonatan', 'age': 30} # Hypothetical age
        describe_user(**user_info) # The ** unpacks the dict into keyword arguments
        ```

  * **Generator Expressions:** Swap brackets for parentheses to create iterators that produce items on demand-fantastic for streaming large files or chaining calculations without materializing intermediate lists.
    ```python
    total = sum(float(row["revenue"]) for row in csv_reader)  # memory stays flat
    ```

  * **`any()` / `all()` / `min()` / `max()` with comprehensions:** These built-ins encapsulate common loops and communicate intent succinctly.
    ```python
    if all(score >= 0 for score in exam_scores):
        ...
    ```

  * **Context Managers (`with`):** Handle setup/teardown of resources-files, database connections, temporary settings-without manual `try/finally` blocks.
    ```python
    with open("metrics.json", "w") as fp:
        json.dump(results, fp)
    ```

  * **`pathlib.Path` instead of raw strings:** It brings operator overloading for path joins (`data_dir / "raw.csv"`) and methods such as `.read_text()` that make file operations expressive and cross-platform.

Treat these idioms as a palette. Pick the one that makes your code both faster *and* easier to understand. Python rewards a blend of elegance and pragmatism.

### Module 1.8: Your Professional Environment

Writing code directly in a plain Python prompt is fine for experiments. Real projects require structure.

  * **Virtual Environments (`venv` or `conda`):**

      * **Problem:** Project A needs `pandas v1.5`, but Project B needs `pandas v2.1`. If you install them globally, one project breaks.
      * **Solution:** A virtual environment creates an isolated Python "sandbox" for each project. Libraries installed inside it do not affect other projects or your system's global Python.
      * **Practice (`venv` - built into Python):**
        1.  `python -m venv .venv` (Creates a folder named `.venv`)
        2.  `source .venv/bin/activate` (Linux/macOS) or `.venv\Scripts\activate` (Windows) (Activates the environment - your prompt changes)
        3.  `pip install pandas numpy scikit-learn` (Installs libraries *only* inside `.venv`)
        4.  `deactivate` (Exits the environment)
      * **Statistician-Pythonista Take:** Non-negotiable for reproducible research and avoiding dependency conflicts.

  * **`.gitignore` File:**

      * **Problem:** You use `git` for version control, but you don't want to commit large data files, virtual environments (`.venv`), or temporary files (`.pyc`).
      * **Solution:** Create a file named `.gitignore` in your project root and list patterns to ignore.
        ```gitignore
        # Virtual Environment
        .venv/

        # Python cache files
        __pycache__/
        *.pyc

        # Data files (if large or sensitive)
        data/*.csv
        *.parquet
        ```

  * **`requirements.txt` File:**

      * **Problem:** You need to share your project or deploy it. How does someone else know exactly which libraries (and versions) to install?
      * **Solution:** A file listing dependencies.
      * **Practice:**
        1.  Activate your virtual environment.
        2.  `pip freeze > requirements.txt` (Generates the file based on installed packages)
        3.  Someone else can then run `pip install -r requirements.txt` to replicate your environment exactly.
      * **Statistician-Pythonista Take:** Essential for reproducibility â€“ a core scientific principle.

  * **Project Layout:** A predictable directory structure keeps collaborators oriented. A common pattern is:
    ```
    project/
    |-- data/            # raw/, processed/, interim/ subfolders
    |-- notebooks/       # exploratory analyses (number them chronologically)
    |-- src/             # reusable modules and pipeline scripts
    |-- tests/           # unit or regression tests
    `-- docs/            # decision logs, user guides, READMEs
    ```
    When every project looks similar, you immediately know where to add code or find results.

  * **Configuration and Secrets:** Store parameters (like file paths or model hyperparameters) in `.yaml` or `.toml` files and load them with libraries such as `pydantic` or `dynaconf`. Never hard-code credentials-use environment variables and tools like `python-dotenv` or secret managers.

  * **Notebooks vs. Scripts:** Use notebooks for exploration and storytelling, but extract reusable logic into modules under `src/`. That allows automated testing, packaging, and reuse across experiments.

  * **Automated Quality Gates:** Linting, formatting, and unit tests can run in a simple `make lint` or `tox` command locally and in CI. Lightweight automation reinforces best practices without manual policing.

Professional environments reward habits that minimize surprises. Setting up these guardrails early lets you focus your creativity on analysis rather than firefighting tooling issues.

    2.  Remove punctuation (you can define a string of punctuation characters to remove or use the `string.punctuation` constant).
    3.  *Optional but recommended:* Split the cleaned line into a list of words.
* **Implementation:** Define a Python function (e.g., `clean_line(text_line)`) with a clear docstring explaining what it does.
* **Deliverable:** The function definition. Test it on a sample line like `"Alice, startled, jumped up!"` and show the output (e.g., `['alice', 'startled', 'jumped', 'up']`).

````python
from string import punctuation


def clean_line(text: str) -> list:
	"""Cleans a line of text by converting to lowercase, removing punctuation, and splitting into words.

	Args:
		text: The input string to clean.

	Returns:
		A list of cleaned words.
	"""
	# Convert to lowercase and remove punctuation, then split into words
	cleaned = text.lower().translate(str.maketrans("", "", punctuation))
	return cleaned.split()
````

**Task 3: Processing Lines Efficiently (Pythonic Idiom)**

* **Objective:** Apply your `clean_line` function to every line in the `raw_text`.
* **Implementation:**
    1.  First, split the `raw_text` into a list of lines.
    2.  Use a **list comprehension** to call `clean_line` on each line. This should produce a list of lists of words.
    3.  *Optional:* Flatten this list of lists into a single list containing all words in the book.
* **Deliverable:** The list comprehension code. Print the first 10 processed words to verify.

````python
# Split the raw_text into lines
lines = raw_text.splitlines()

# Use list comprehension to apply clean_line to each line, producing a list of lists
processed_lines = [clean_line(line) for line in lines]

# Flatten the list of lists into a single list of all words
all_words = [word for line_words in processed_lines for word in line_words]

print(all_words[:10])
````

````python
from collections import Counter

# This iterates through the list and counts occurrences of each unique word
word_count = Counter(all_words)

# Call most_common(10) on the Counter to get the 10 most frequent words as a list of tuples
print(word_count.most_common(10))
````
