## Module 0.1 - The Grammar of Statistical Notation

### Deconstructing the Core Symbols

#### 1. Introduction to Mathematical Shorthand

Mathematical and statistical notation can appear intimidating. However, it is simply a language designed for maximum precision and minimum ambiguity. Think of it not as a complex calculation, but as a highly efficient form of shorthand. Our goal in this section is to learn the "alphabet" and "basic vocabulary" of this language. Once you master these, formulas will transform from cryptic instructions into readable sentences.

We will use a simple dataset for all our examples in this module:
`my_data = [4, 8, 5, 10, 3]`


#### 2. The Basic Nouns: Representing Data

##### **The Symbol: `x`**
* **Pronunciation:** "x"
* **Linguistic Role:** A **general noun**. It represents the concept or variable we are measuring.
* **Definition:** `x` refers to our variable of interest. If our data represents the number of hours studied, `x` is "hours studied."

##### **The Symbol: `i`**
* **Pronunciation:** "i"
* **Linguistic Role:** A **counter or index**. It helps us point to a specific item in a sequence.
* **Definition:** `i` is a placeholder for the position of a value in our dataset (1st, 2nd, 3rd, etc.).

##### **The Symbol: `xᵢ`**
* **Pronunciation:** "x sub i"
* **Linguistic Role:** A **specific pronoun** like "this one" or "that one." It combines `x` and `i` to refer to a single, specific data point.
* **Definition:** `xᵢ` refers to the *i*-th value in our dataset `x`.
* **Example with `my_data`:**
    * `x₁` is the 1st value, which is **4**.
    * `x₃` is the 3rd value, which is **5**.
    * `x₅` is the 5th value, which is **3**.

##### **The Symbol: `N`**
* **Pronunciation:** "capital N"
* **Linguistic Role:** A **quantifier**. It tells us "how many."
* **Definition:** `N` represents the total number of observations in our dataset.
* **Example with `my_data`:**
    * `N = 5`, because there are five values in the list.

#### 3. The Main Verb: Summation

##### **The Symbol: `Σ`**
* **Pronunciation:** "capital sigma"
* **Linguistic Role:** An **action verb**. It means "sum them all up."
* **Definition:** This is the summation operator. It is the mathematical equivalent of a `for` loop that iterates through a sequence and adds up the elements.
* **Example:** The expression `Σ xᵢ` is a full command. It translates to: "Iterate through all values of `x` from the first (`i=1`) to the last (`i=N`) and add them all together."
* **Example with `my_data`:**
    * `Σ xᵢ = x₁ + x₂ + x₃ + x₄ + x₅`
    * `Σ xᵢ = 4 + 8 + 5 + 10 + 3 = 30`

#### 4. Reading Our First "Sentence"

Let's introduce one more symbol, which we will deconstruct fully in the next module.

##### **The Symbol: `μ`**
* **Pronunciation:** "mu"
* **Definition:** This symbol represents the **population mean**, or average.

Now, we can combine our new vocabulary to form a complete, meaningful sentence. The formula for the mean is:

$$ \mu = \frac{\sum x_i}{N} $$

Let's translate this sentence, piece by piece:
* `μ`: "The mean is..."
* `=` : "...equal to..."
* `Σ xᵢ`: "...the sum of all the data points..."
* `/ N`: "...divided by the total number of data points."

**Full Translation:** "The mean is equal to the sum of all the data points divided by the total number of data points."



```python
my_data = [4, 8, 5, 10, 3]
```

```python
def mean(numbers: list) -> float:
    return sum(numbers) / len(numbers)
```

```python
mean(numbers=my_data)
```

## Module 1.1: The Mean - The Center of Gravity

### 1. Conceptual Overview

The **mean**, or arithmetic average, is the most common measure of central tendency. Conceptually, it represents the "balance point" of a dataset. If you were to place each data point as an equal weight on a beam, the mean would be the fulcrum's position where the beam balances perfectly. It gives us a single value that summarizes the center of our data.

### 2. The Formula (Recap)

As established in Module 0.1, the formula for the population mean (`μ`) is:

$$ \mu = \frac{\sum x_i}{N} $$

In plain English, this reads: "The mean is the sum of all the data points, divided by the number of data points."

### 3. Implementation from First Principles

To fully internalize the formula, we will translate it into a Python function without using the built-in `sum()` function. This exercise forces us to manually implement the `Σ` (summation) operator.

The logic is as follows:
1.  Initialize a variable, `total`, to `0.0` to hold our running sum.
2.  Get the number of items, `N`, from the length of the list.
3.  Loop through each `number` in the list.
4.  In each loop, add the current `number` to our `total`.
5.  After the loop, divide the final `total` by `N`.

```python
def calculate_mean_verbose(data: list[float | int]) -> float:
    """
    Calculates the arithmetic mean (average) of a list of numbers from first principles.

    The mean represents the "balance point" or center of gravity of the dataset.
    It is computed as the sum of all data points divided by the total number of points (N).
    This function implements the summation (Σ) manually using a loop, without relying on built-in sum().

    Parameters
    ----------
    data : list[float | int]
            A list of numerical values (floats or ints) representing the dataset.
            For example, [4, 8, 5, 10, 3].

    Returns
    -------
    float
            The arithmetic mean of the data. Returns 0.0 if the list is empty to avoid division by zero.
    """
    # Get the total number of observations (N) in the dataset
    n = len(data)

    # Handle edge case: if no data points, return 0.0 (though in practice, mean of empty set is undefined)
    if n == 0:
        return 0.0

    # Initialize total_sum to 0.0; this will accumulate the sum of all x_i (data points)
    total_sum = 0.0

    # Manually implement the summation (Σ): loop through each data point and add to total_sum
    # This is equivalent to Σ x_i from i=1 to N
    for number in data:
        total_sum = total_sum + number  # Add each x_i to the running total

    # Compute the mean: μ = Σ x_i / N
    mean_value = total_sum / n
    return mean_value
```

```python
def calculate_mean_pythonic(data: list[float | int]) -> float:
    """
    Calculates the arithmetic mean (average) of a list of numbers using Python's built-in functions.

    The mean represents the "balance point" or center of gravity of the dataset.
    It is computed as the sum of all data points divided by the total number of points (N).

    Parameters
    ----------
    data : list[float | int]
            A list of numerical values (floats or ints) representing the dataset.
            For example, [4, 8, 5, 10, 3].

    Returns
    -------
    float
            The arithmetic mean of the data. Returns 0.0 if the list is empty to avoid division by zero.
    """
    n = len(data)
    if n == 0:
        return 0.0

    return sum(data) / n
```

## Module 1.2: The Median - The Robust Midpoint

### 1. Conceptual Overview

The **median** is the value that separates the higher half from the lower half of a dataset. To find it, you must first sort the data. The median is then the physical **middle value**.

Unlike the mean, which is calculated from every value in the dataset, the median's value depends only on the one or two data points in the very center of the sorted list. This makes it exceptionally **robust** (resistant) to outliers.

Let's revisit our real estate example: A mansion is built in a neighborhood of modest homes.
* The **mean** house price would skyrocket, giving a misleading impression of the area's affordability.
* The **median** house price would barely change, as the new mansion is just one new data point at the extreme high end, and the "middle" house in the sorted list remains largely the same.

Therefore, for skewed datasets (like income, house prices, or internet latency), the median is often a far more honest and representative measure of the "typical" value.

### 2. The Algorithm: An Ordered Process

The "formula" for the median is actually an algorithm:

1.  **Sort** the dataset in ascending order.
2.  Determine the number of data points, `N`.
3.  Apply one of the following rules based on whether `N` is odd or even:
    * **If N is odd:** The median is the single value at the center. Its position is `(N + 1) / 2`. For a list of 9 items, the median is the `(9 + 1) / 2 = 5th` value.
    * **If N is even:** There is no single middle value. The median is the **mean of the two central values**. Their positions are `N / 2` and `(N / 2) + 1`. For a list of 10 items, the median is the mean of the `10 / 2 = 5th` and `(10 / 2) + 1 = 6th` values.

### 3. Implementation from First Principles

Translating this algorithm into Python requires careful handling of list indexing, which starts at 0.

* The `sorted()` function is ideal here because it returns a new, sorted list without modifying the original data.
* We use the modulo operator (`%`) to check if `N` is even (`n % 2 == 0`).
* We must translate from the 1-based positions in the algorithm to 0-based list indices. For a list of 5 items (`n=5`), the middle position is the 3rd, but its index is `2`. The formula for the index of the middle element in an odd-sized list is `(n - 1) // 2`.


```python
def calculate_median(data: list[float | int]) -> float:
    """
    Calculates the median of a list of numbers from first principles.

    The median is the value that separates the higher half from the lower half
    of a dataset. It is computed by sorting the data and finding the middle value(s).

    Parameters
    ----------
    data : list[float | int]
            A list of numerical values (floats or ints) representing the dataset.
            For example, [4, 8, 5, 10, 3].

    Returns
    -------
    float
            The median of the data. Returns 0.0 if the list is empty.
    """
    n = len(data)
    if n == 0:
        return 0.0

    # Step 1: Sort the data.
    sorted_data = sorted(data)

    # Step 2: Check if N is odd or even.
    if n % 2 == 1:
        # N is odd.
        # Find the index of the middle element.
        middle_index = (n - 1) // 2
        median = sorted_data[middle_index]
    else:
        # N is even.
        # Find the indices of the two middle elements.
        upper_middle_index = n // 2
        lower_middle_index = upper_middle_index - 1

        # The median is the mean of the two middle elements.
        median = (sorted_data[lower_middle_index] + sorted_data[upper_middle_index]) / 2

    return float(median)
```
