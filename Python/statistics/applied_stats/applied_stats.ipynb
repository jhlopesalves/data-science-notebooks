
#### Module 1.1: The Core Concepts of Probability

This section covers the fundamental definitions used to describe random phenomena.

##### **1. Experiment**

An **experiment** is any procedure that yields one of a set of possible outcomes. It is characterized by the fact that its outcome is not known in advance, but the set of all possible outcomes is known.

  * **Example 1:** Tossing a coin.
  * **Example 2:** Rolling a six-sided die.
  * **Example 3:** Measuring the time it takes for a web page to load.

##### **2. Sample Space ($S$)**

The **sample space** of an experiment, denoted by $S$ or $\Omega$, is the set of all possible distinct outcomes of that experiment.

  * **For a coin toss:** $S = \{Heads, Tails\}$
  * **For a die roll:** $S = \{1, 2, 3, 4, 5, 6\}$

The sample space must be **mutually exclusive** (only one outcome can occur per trial) and **collectively exhaustive** (it includes all possible outcomes).

##### **3. Event ($A$)**

An **event** is any subset of the sample space $S$. It represents a specific outcome or a collection of outcomes of interest. Events are typically denoted by capital letters (e.g., $A, B, C$).

  * **Experiment:** Rolling a die ($S = \{1, 2, 3, 4, 5, 6\}$).
  * **Event A:** The outcome is an even number.
      * $A = \{2, 4, 6\}$
  * **Event B:** The outcome is less than 3.
      * $B = \{1, 2\}$

##### **4. Probability of an Event**

The probability of an event $A$, denoted $P(A)$, is a value between 0 and 1 that measures the likelihood of that event occurring.

  * $P(A) = 0$ means event $A$ is impossible.
  * $P(A) = 1$ means event $A$ is certain.

For experiments with a finite number of **equally likely outcomes**, the probability of an event $A$ is calculated as the ratio of the number of outcomes in $A$ to the total number of outcomes in the sample space $S$. Let $|A|$ denote the number of elements in event $A$ and $|S|$ denote the number of elements in the sample space.

$$P(A) = \frac{|A|}{|S|}$$

  * **Example:** For a fair die roll, what is the probability of rolling an even number (Event $A$)?
      * $S = \{1, 2, 3, 4, 5, 6\}$, so $|S| = 6$.
      * $A = \{2, 4, 6\}$, so $|A| = 3$.
      * $P(A) = \frac{3}{6} = 0.5$
