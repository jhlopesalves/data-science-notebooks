**1. Order of Operations (PEMDAS/BODMAS): The Rules of the Road**

Imagine you see the expression: $3 + 4 \times 2$. How do you calculate it?
* Do you add first? $3 + 4 = 7$, then $7 \times 2 = 14$.
* Or do you multiply first? $4 \times 2 = 8$, then $3 + 8 = 11$.

Mathematics needs a single, unambiguous answer. The **Order of Operations** provides this rulebook. It's a convention everyone agrees on so that expressions have only one correct interpretation. You might know it by acronyms like PEMDAS or BODMAS.

**The Hierarchy:**

1.  **P**arentheses (or **B**rackets): Operations inside parentheses `()` or brackets `[]` are performed first, working from the innermost pair outwards.
2.  **E**xponents (or **O**rders): Next, calculate powers and roots (e.g., $x^2$, $\sqrt{x}$).
3.  **M**ultiplication and **D**ivision: Perform these next, working from left to right. They have equal precedence.
4.  **A**ddition and **S**ubtraction: Perform these last, working from left to right. They also have equal precedence.

**Why it's Crucial:** Without this convention, formulas would be ambiguous and useless. It ensures that $3 + 4 \times 2$ *always* equals 11, never 14.

**Step-by-Step Example:** Evaluate $10 - (2 + 1) \times 3^2 \div 9$.

1.  **Parentheses:** Calculate $(2 + 1) = 3$.
    * Expression becomes: $10 - 3 \times 3^2 \div 9$.
2.  **Exponents:** Calculate $3^2 = 9$.
    * Expression becomes: $10 - 3 \times 9 \div 9$.
3.  **Multiplication and Division (Left to Right):**
    * First, $3 \times 9 = 27$.
        * Expression becomes: $10 - 27 \div 9$.
    * Next, $27 \div 9 = 3$.
        * Expression becomes: $10 - 3$.
4.  **Addition and Subtraction (Left to Right):**
    * Finally, $10 - 3 = 7$.

**The unambiguous result is 7.** Every calculation in data science, from simple averages to complex algorithms, relies implicitly on these rules.

**2. Variables: Placeholders for Quantities**

In mathematics and programming, we often use letters (like $x$, $y$, $a$, $\beta$, etc.) to represent quantities. These are called **variables**. A variable can represent:

* **An Unknown Value:** In algebra, we solve equations like $2x + 3 = 11$ to find the specific value of $x$ that makes the statement true (here, $x=4$).
* **A Changing Quantity:** In functions, like $f(x) = x^2$, the variable $x$ can take on *many* different values (the inputs).
* **A Parameter:** In models, letters often represent fixed but perhaps unknown constants that define the model's structure (e.g., the slope $\beta_1$ in linear regression).

**Why Variables are Essential:** They allow us to express general relationships and rules without needing to specify exact numbers immediately. The formula for the area of a rectangle, $A = l \times w$, works for *any* length $l$ and width $w$. Variables give mathematics its power of abstraction.

**3. Functions ($f(x)$): The Input-Output Machine**

A **function** is a specific mathematical rule that takes an **input**, performs some operation on it, and produces exactly one **output**. Think of it like a machine : you put something in, the machine does something according to its design, and you get one specific thing out.

**Notation:** We usually write functions as $f(x)$, read "f of x".
* $f$: The name of the function (the "machine's name").
* $x$: The input variable (the "material you put in").
* $f(x)$: The output value produced when the rule $f$ is applied to the input $x$ (the "product that comes out").

**Example:** $f(x) = x^2$
* **Rule:** Square the input.
* If the input is $x=3$, the output is $f(3) = 3^2 = 9$.
* If the input is $x=-2$, the output is $f(-2) = (-2)^2 = 4$.

**Key Concepts:**
* **Domain:** The set of all possible valid inputs ($x$ values) for the function. For $f(x) = x^2$, the domain is usually all real numbers ($\mathbb{R}$). For $g(x) = \sqrt{x}$, the domain is non-negative real numbers ($x \ge 0$).
* **Range (or Codomain):** The set of all possible output values ($f(x)$ values) the function can produce. For $f(x) = x^2$, the range is non-negative real numbers ($f(x) \ge 0$).
* **Independent Variable:** The input variable (usually $x$). You choose its value freely (within the domain).
* **Dependent Variable:** The output variable (often written as $y = f(x)$). Its value *depends* on the input chosen.

**Why Functions are Central to Data Science:** Models *are* functions. A regression model is a function that takes features (inputs) and outputs a prediction. A classification model is a function that takes features and outputs a class label or probability. Understanding function notation ($y = f(X, \beta)$) is fundamental to reading any modeling literature.

**4. Summation ($\Sigma$) and Product ($\Pi$) Notation: Simplifying Repetition**

Often, we need to add or multiply many terms that follow a pattern. Writing them all out is tedious and sometimes impossible. Summation and product notation provide a compact way to express these operations.

**Summation Notation ($\Sigma$):** The Greek capital letter Sigma ($\Sigma$) means "sum up."

$$\sum_{i=m}^{n} a_i = a_m + a_{m+1} + \dots + a_n$$

* $\Sigma$: The summation symbol.
* $i$: The **index** of summation (a temporary variable).
* $m$: The **lower limit** (the starting value for $i$).
* $n$: The **upper limit** (the ending value for $i$).
* $a_i$: The expression to be summed. It usually depends on the index $i$.

**Step-by-Step Expansion:** Evaluate $\sum_{i=1}^{4} i^2$.

1.  **Start index:** $i=1$. Term is $1^2 = 1$.
2.  **Next index:** $i=2$. Term is $2^2 = 4$.
3.  **Next index:** $i=3$. Term is $3^2 = 9$.
4.  **End index:** $i=4$. Term is $4^2 = 16$.
5.  **Sum them up:** $1 + 4 + 9 + 16 = 30$.
    * Therefore, $\sum_{i=1}^{4} i^2 = 30$.

**Product Notation ($\Pi$):** The Greek capital letter Pi ($\Pi$) means "multiply."

$$\prod_{i=m}^{n} a_i = a_m \times a_{m+1} \times \dots \times a_n$$

The notation works exactly the same way as summation, but you multiply the terms instead of adding them.

**Step-by-Step Expansion:** Evaluate $\prod_{i=1}^{3} (i+1)$.

1.  **Start index:** $i=1$. Term is $(1+1) = 2$.
2.  **Next index:** $i=2$. Term is $(2+1) = 3$.
3.  **End index:** $i=3$. Term is $(3+1) = 4$.
4.  **Multiply them:** $2 \times 3 \times 4 = 24$.
    * Therefore, $\prod_{i=1}^{3} (i+1) = 24$.

**Why These Notations are Crucial:** They appear *everywhere* in statistics and machine learning.
* The formula for the mean (average): $\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$.
* The Sum of Squared Errors (SSE) in regression: $SSE = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$.
* Likelihood functions in statistics often involve products ($\Pi$).

Now we move from the chalkboard to the keyboard. How do these abstract rules and notations look in Python code?

**1. Order of Operations in Python**

Python's built-in mathematical operators (`+`, `-`, `*`, `/`, `**` for exponentiation, `//` for integer division, `%` for modulo) strictly follow the standard Order of Operations (PEMDAS/BODMAS). Parentheses `()` are used exactly as in mathematics to override the default order.

Let's use our Phase 1 example: $10 - (2 + 1) \times 3^2 \div 9$. In Python, this translates directly:

```python
# Phase 1 calculation: 10 - (2 + 1) * 3**2 / 9
result = 10 - (2 + 1) * 3**2 / 9

# Note: In Python 3, / performs true division (resulting in a float)
# If integer division is intended (like in some contexts), use //

print(f"The result is: {result}")
```

**Step-by-step breakdown (how Python evaluates):**

1.  Parentheses: `(2 + 1)` becomes `3`.
      * Expression: `10 - 3 * 3**2 / 9`
2.  Exponentiation: `3**2` becomes `9`.
      * Expression: `10 - 3 * 9 / 9`
3.  Multiplication and Division (Left to Right):
      * `3 * 9` becomes `27`.
          * Expression: `10 - 27 / 9`
      * `27 / 9` becomes `3.0` (note the float).
          * Expression: `10 - 3.0`
4.  Subtraction: `10 - 3.0` becomes `7.0`.

Python gets the same unambiguous result (7.0) because it follows the same rules we discussed. You rarely need to think about PEMDAS explicitly, but it's crucial to know Python enforces it correctly.

**2. Variables in Python**

In Python, variables are created using the assignment operator (`=`). You give a name (following Python's naming rules) to a value.

```python
# Assigning values to variables
length = 10
width = 5

# Using variables in calculations
area = length * width

print(f"Length: {length}")
print(f"Width: {width}")
print(f"Area: {area}")

# Variables can be updated
length = 12
new_area = length * width
print(f"Updated Length: {length}")
print(f"New Area: {new_area}")
```


Variables in Python work just like their mathematical counterparts: they are names that hold values, allowing for general calculations and easy updates.

**3. Functions in Python**

We define functions in Python using the `def` keyword. They take inputs (arguments or parameters) and return an output using the `return` statement.

Let's implement our $f(x) = x^2$ example:

```python
def square(x):
	"""Calculates the square of the input number."""
	# The rule: square the input
	output = x**2
	return output
```

```python
# Using the function
input_value = 3
output_value = square(input_value)
print(f"The square of {input_value} is {output_value}")

input_value = -2
output_value = square(input_value)
print(f"The square of {input_value} is {output_value}")
```

Python functions directly map to the mathematical concept: a named rule that transforms inputs into a single, well-defined output.


**Lambda Functions (A Quick Preview):**
Python also allows for small, anonymous functions using `lambda`:


```python
print(f"Using lambda: square of 5 is {(lambda x: x**2)(5)}")
```

We will revisit `lambda` functions later, as they are very common in data manipulation with Pandas.



**4. Summation ($\Sigma$) and Product ($\Pi$) in Python**

While Python doesn't have a direct $\Sigma$ or $\Pi$ symbol, we can implement these operations using loops or built-in functions, especially with NumPy for efficiency on numerical arrays.

**Summation ($\Sigma$):**

  * **Using a `for` loop (general, works on any iterable):**
    Let's calculate $\sum_{i=1}^{4} i^2$.

```python
total = 0
for i in range(1, 5):
	term = i**2
	total = total + term
	print(f"i={i}, term={term}, current total={total}")
print(f"Final Sum (loop): {total}")
```

  * **Using Python's built-in `sum()` with a generator expression:**
    More concise for simple sums.


```python
total_sum = sum(i**2 for i in range(1, 5))
print(f"Final Sum (sum()): {total_sum}")
```


  * **Using NumPy (highly efficient for numerical arrays):**
    This is the standard approach in data science when working with numerical data.


```python
total_np_sum = np.sum(np.arange(1, 5) ** 2)
print(f"Final Sum (NumPy): {total_np_sum}")
```



    Notice how NumPy avoids explicit Python loops, which is much faster for large arrays.

**Product ($\Pi$):**

  * **Using a `for` loop:**
    Let's calculate $\prod_{i=1}^{3} (i+1)$.

```python
product_val = 1
for i in range(1, 4):
	term = i + 1
	product_val = product_val * term
	print(f"i={i}, term={term}, current product={product_val}")
print(f"Final Product (loop): {product_val}")
```

  * **Using NumPy:**

```python
# Terms are (1+1), (2+1), (3+1) -> 2, 3, 4
terms = np.arange(1, 4) + 1

# Calculate the product
total_np_product = np.prod(terms)
# Or simply: total_np_product = np.prod(np.arange(1, 4) + 1)

print(f"NumPy array of terms: {terms}")
print(f"Final Product (NumPy): {total_np_product}")
```

**Key Takeaway for $\Sigma$ and $\Pi$:**
While standard Python loops work, **NumPy's `np.sum()` and `np.prod()` are the preferred tools** when dealing with numerical sequences or arrays in data science due to their speed and conciseness, stemming from *vectorization*.
