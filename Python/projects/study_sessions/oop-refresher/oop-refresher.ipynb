#### Mission A1 – The absolute minimum class
Write the smallest possible class that does something useful when instantiated.

Requirements:
- Class name: `Point`
- Takes `x` and `y` in `__init__`
- Stores them as instance attributes

#### Mission A2 – Add a readable representation
Add a `__repr__` method so that when you type `p = Point(3, 4)` and then just `p` in a REPL you see something useful like `Point(x=3, y=4)` instead of `<__main__.Point object at 0x…>`.

#### Mission A3 – Add behavior
Add an instance method called `distance_from_origin(self)` that returns the Euclidean distance from (0, 0).  
Do NOT use `math.sqrt` yet — just return the squared distance (we’ll improve it later.

#### Mission A4 – `@property` in action
Turn `distance_from_origin` into a read-only property called `distance` (the user writes `p.distance`, not `p.distance()`).


#### Mission A5 – One static method and one class method
Add:
- A static method `from_polar(r, theta)` that returns a new `Point` from polar coordinates (use `math.radians` if you want, or just assume degrees/no import)
- A class method `zero()` that returns Point(0, 0)



```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self, x, y):
        return f"Point(x={self.x}, y={self.y})"

    def distance_from_origin(self):
        return (self.x**2, self.y**2)
```
