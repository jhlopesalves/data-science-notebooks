### Spearman Correlation Formula

The Spearman correlation coefficient (ρ) measures the monotonic relationship between two variables. It is calculated as:

$$
\rho = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}
$$

Where:
- $d_i$ = difference between the ranks of corresponding variables
- $n$ = number of observations

## Interpretation
- **ρ = +1**: Perfect positive monotonic relationship (as one variable increases, the other always increases)
- **ρ = -1**: Perfect negative monotonic relationship (as one variable increases, the other always decreases)  
- **ρ = 0**: No monotonic relationship between variables

Unlike Pearson correlation which measures linear relationships, Spearman correlation captures any monotonic relationship (linear or nonlinear) by using rank orders rather than raw values, making it more robust to outliers.

```python
cols = ["OverallQual", "OverallCond", "YearBuilt", "GarageMissing"]
corr_spearman = X[cols].corr(method="spearman")

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(
	corr_spearman,
	annot=True,
	cmap="RdBu_r",
	annot_kws={"size": 10},
	fmt=".2f",
)
ax.set_title("Spearman Correlation Heatmap", fontsize=14)
ax.set_xlabel("")
ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

### Point-Biserial Correlation Formula

The point-biserial correlation coefficient ($r_{pb}$) measures the relationship between a continuous variable and a dichotomous variable. It is calculated as:

$$
r_{pb} = \frac{M_1 - M_0}{s} \cdot \sqrt{p(1-p)}
$$

Where:
- $M_1$ = mean of the continuous variable for group 1 (coded as 1)
- $M_0$ = mean of the continuous variable for group 0 (coded as 0)
- $s$ = standard deviation of the continuous variable
- $p$ = proportion of cases in group 1

## Interpretation
- **$r_{pb} = +1$**: Perfect positive relationship (all values in group 1 are higher than group 0)
- **$r_{pb} = -1$**: Perfect negative relationship (all values in group 1 are lower than group 0)
- **$r_{pb} = 0$**: No relationship between the dichotomous and continuous variables
