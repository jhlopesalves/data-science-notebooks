# Ames Housing Dataset


## Overview

This dataset contains comprehensive information about residential homes in Ames, Iowa. The dataset includes 79 explanatory variables describing almost every aspect of residential properties, challenging analysts to predict the final sale price of each home.

## Dataset Details

- **Source**: Kaggle Competition - House Prices: Advanced Regression Techniques
- **Format**: ARFF (Attribute-Relation File Format)
- **Target Variable**: SalePrice
- **Records**: Multiple residential properties in Ames, Iowa
- **OpenML ID**: 42165

### Dropping, Mapping, Encodding and Filling

```python
X_train = X_train.drop(columns=["PoolQC", "MiscFeature", "Alley", "Fence"])
X_test = X_test.drop(columns=["PoolQC", "MiscFeature", "Alley", "Fence"])
```

```python
categorical = X_train.select_dtypes("object").columns.to_list()
numeric = X_train.select_dtypes("number").columns.to_list()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

# Categorical columns where missing = "None"
cat_absence_cols = [
    "GarageCond",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "BsmtFinType2",
    "BsmtExposure",
    "BsmtCond",
    "BsmtQual",
    "BsmtFinType1",
    "FireplaceQu",
    "MasVnrType",
]

# Numeric columns where missing = 0
num_absence_cols = ["MasVnrArea", "GarageYrBlt"]

# Numeric columns needing median imputation
median_impute_cols = ["LotFrontage"]

# Categorical columns needing mode imputation
mode_impute_cols = ["Electrical"]

cat_absence_imputer = SimpleImputer(strategy="constant", fill_value="None")
num_absence_imputer = SimpleImputer(strategy="constant", fill_value=0)
median_imputer = SimpleImputer(strategy="median")
mode_imputer = SimpleImputer(strategy="most_frequent")

# The actual impute transformer
imputation = ColumnTransformer(
    [
        ("cat_absence_imputer", cat_absence_imputer, cat_absence_cols),
        ("num_absence_imputer", num_absence_imputer, num_absence_cols),
        ("median_imputer", median_imputer, median_impute_cols),
        ("mode_imputer", mode_imputer, mode_impute_cols),
    ],
    remainder="passthrough",
)
```

```python
imputation.fit(X_train)
```
