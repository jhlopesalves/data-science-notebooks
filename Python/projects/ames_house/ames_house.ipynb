
### Log-Transformed Sale Price

The `SalePriceLog` variable is created by applying a natural logarithm transformation to the original sale prices. 

#### Why Log Transform Sale Prices?

1. **Normalizes Skewed Distribution**: Housing prices typically follow a right-skewed distribution. Log transformation helps make the distribution more normal, which is beneficial for many statistical methods and machine learning algorithms.

2. **Stabilizes Variance**: The variance of housing prices often increases with the mean. Log transformation stabilizes the variance across different price ranges.

3. **Interpretable as Percentage Changes**: After log transformation:
	- A difference of 1 unit represents a ~100% change in price
	- A difference of 0.1 units represents ~10% change in price
	- This makes percentage-based comparisons more intuitive

4. **Reduces Heteroscedasticity**: Helps meet the assumption of constant variance in linear regression models.

```python
X["SalePriceLog"] = np.log1p(y)
```

### Point-biserial correlation

The point-biserial correlation coefficient $r_{pb}$ measures the association between
a **dichotomous** variable (e.g. 0/1, yes/no) and a **continuous** variable. It is
mathematically equivalent to Pearson's correlation when one of the variables is binary.

The formula is:

$$
r_{pb} = \frac{\mu_1 - \mu_0}{\sigma} \cdot \sqrt{\frac{n_0 n_1}{n^2}}
$$

Where:

- $\mu_1$ - mean of the continuous variable for group 1 (binary value = 1)
- $\mu_0$ - mean of the continuous variable for group 0 (binary value = 0)
- $\sigma$ - sample standard deviation of the continuous variable (all observations)
- $n_1$ - number of observations with binary value = 1
- $n_0$ - number of observations with binary value = 0
- $n = n_0 + n_1$ - total number of observations

So $r_{pb}$ is literally:  

> **(difference in group means)**, normalized by  
> **overall spread** and **group balance**.

#### Interpretation

- $r_{pb} = +1$ - perfect positive separation  
  (all values in group 1 are higher than all values in group 0)
- $r_{pb} = -1$ - perfect negative separation  
  (all values in group 1 are lower than all values in group 0)
- $r_{pb} = 0$ - no linear/monotonic relationship between the groups and the continuous variable

Effect size heuristics (Cohen-style, rough):

- $|r_{pb}| \approx 0.10$ - small
- $|r_{pb}| \approx 0.30$ - medium
- $|r_{pb}| \approx 0.50$ - large

Typical use:

- Binary indicator (e.g. `GarageMissing`, treatment vs control, male/female)
- Continuous outcome (e.g. price, score, area)

This quantity is closely related to the **two-sample t-test**: a large magnitude $|r_{pb}|$
corresponds to a large difference in means between the two groups.

```python
def point_biserial_corr(
	df: pd.DataFrame,
	binary_col: str,
	continuous_col: str,
) -> float:
	"""
	Calculate the point-biserial correlation coefficient between a binary and continuous variable.

	The point-biserial correlation measures the strength and direction of association between
	a dichotomous variable and a continuous variable. It is mathematically equivalent to the
	Pearson correlation when one variable is binary.

	## Theoretical Background

	The point-biserial correlation coefficient (r_pb) measures the association between
	a **dichotomous** variable (e.g., 0/1, yes/no) and a **continuous** variable. It is
	calculated using the formula:

	r_pb = [(μ₁ - μ₀) / σ] × √[(n₀ × n₁) / n²]

	Where:
	- μ₁ = mean of the continuous variable for group 1 (binary value = 1)
	- μ₀ = mean of the continuous variable for group 0 (binary value = 0)
	- σ = sample standard deviation of the continuous variable (all observations)
	- n₁ = number of observations with binary value = 1
	- n₀ = number of observations with binary value = 0
	- n = n₀ + n₁ = total number of observations

	Essentially, r_pb represents the **difference in group means** normalized by the 
	**overall spread** and **group balance**.

	## Interpretation

	- **r_pb = +1**: Perfect positive separation (all values in group 1 are higher than all values in group 0)
	- **r_pb = -1**: Perfect negative separation (all values in group 1 are lower than all values in group 0)
	- **r_pb = 0**: No linear/monotonic relationship between the groups and the continuous variable

	## Effect Size Guidelines (Cohen, 1988)

	- **|r_pb| ≈ 0.10**: Small effect
	- **|r_pb| ≈ 0.30**: Medium effect
	- **|r_pb| ≈ 0.50**: Large effect

	Parameters
	----------
	df : pd.DataFrame
		DataFrame containing the variables of interest.
	binary_col : str
		Name of the binary column (must contain only 0, 1, True, or False).
	continuous_col : str
		Name of the continuous column.

	Returns
	-------
	float
		Point-biserial correlation coefficient in [-1, 1].

	Raises
	------
	ValueError
		- If there are fewer than 3 complete observations.
		- If binary_col contains values other than 0, 1, True, or False.
		- If either group (0 or 1) has no observations.
		- If the continuous variable has zero variance.

	Examples
	--------
	>>> import pandas as pd
	>>> import numpy as np
	>>> 
	>>> # Example 1: Strong positive correlation
	>>> df1 = pd.DataFrame({
	...     'GarageMissing': [0, 0, 0, 1, 1, 1],
	...     'SalePrice': [250000, 275000, 300000, 150000, 175000, 200000]
	... })
	>>> r_pb = point_biserial_corr(df1, 'GarageMissing', 'SalePrice')
	>>> print(f"Point-biserial correlation: {r_pb:.3f}")
	Point-biserial correlation: -1.000  # Perfect negative correlation
	>>>
	>>> # Example 2: Weak correlation
	>>> df2 = pd.DataFrame({
	...     'GarageMissing': [0, 0, 1, 1, 0, 1],
	...     'LotArea': [5000, 6000, 5500, 5200, 5800, 5100]
	... })
	>>> r_pb = point_biserial_corr(df2, 'GarageMissing', 'LotArea')
	>>> print(f"Point-biserial correlation: {r_pb:.3f}")
	Point-biserial correlation: -0.245  # Weak negative correlation
	>>>
	>>> # Example 3: Real-world Ames housing data
	>>> # This would show the correlation between garage missingness and sale price
	>>> # r_pb = point_biserial_corr(X, 'GarageMissing', 'SalePriceLog')
	"""
	data = df[[binary_col, continuous_col]].dropna()

	if data.shape[0] < 3:
		raise ValueError("At least 3 complete observations are required.")

	x = data[binary_col]
	y = data[continuous_col].astype(float)

	# Check that the binary column contains only allowed values
	unique_vals = set(np.unique(x))
	if not unique_vals.issubset({0, 1, True, False}):
		raise ValueError("binary_col must contain only 0, 1, True, or False.")

	# Convert boolean values to 0/1 for consistent processing
	x = x.replace({True: 1, False: 0})

	# Extract groups
	y_values = y.to_numpy()
	group_0 = y_values[x == 0]
	group_1 = y_values[x == 1]

	n_0, n_1 = len(group_0), len(group_1)
	if n_0 == 0 or n_1 == 0:
		raise ValueError("Both groups (0 and 1) must have at least one observation.")

	# Sample standard deviation (ddof = 1)
	sigma = np.std(y_values, ddof=1)
	if sigma <= 0:
		raise ValueError("Variance of continuous_col is zero or negative.")

	mu_0 = np.mean(group_0)
	mu_1 = np.mean(group_1)
	n = len(y_values)

	pb = (mu_1 - mu_0) / sigma * np.sqrt((n_0 * n_1) / n**2)
	return float(pb)
```

### Cramér's V

Cramér's V measures the strength of association between two **categorical** variables.
It is based on the chi-squared statistic from a contingency table and scaled to lie
between 0 and 1.

Start with the chi-squared statistic:

$$
\chi^2 = \sum_{i,j} \frac{(O_{ij} - E_{ij})^2}{E_{ij}}
$$

Where:

- $O_{ij}$ - observed counts in cell $(i, j)$
- $E_{ij}$ - expected counts under independence

Define:

$$
\phi^2 = \frac{\chi^2}{n}
$$

Then Cramér's V is:

$$
V = \sqrt{ \frac{\phi^2}{\min(r - 1, k - 1)} }
     = \sqrt{ \frac{\chi^2}{n \cdot \min(r - 1, k - 1)} }
$$

Where:

- $n$ - total number of observations
- $r$ - number of rows (categories of one variable)
- $k$ - number of columns (categories of the other variable)
- $\min(r - 1, k - 1)$ - normalisation by the smaller degrees of freedom

#### Interpretation

- $V = 0$ - no association (variables are independent)
- $V = 1$ - perfect association
- $0 < V < 1$ - increasing strength of association as $V$ approaches 1

Typical effect size heuristics (very rough):

- $V \approx 0.10$ - small association
- $V \approx 0.30$ - medium association
- $V \approx 0.50$ - large association

Cramér's V is:

- symmetric: $V(X, Y) = V(Y, X)$
- normalised: values are comparable across different contingency table sizes
- useful for quantifying relationships like **GarageMissing × Neighborhood**,  
  **GarageMissing × MSSubClass**, etc., when analysing structured missingness.

```python
def cramersv_corr(df: pd.DataFrame, col_1: str, col_2: str) -> float:
	"""
	Calculate Cramér's V coefficient between two categorical variables.

	Cramér's V is a measure of association between nominal variables, providing
	a standardized effect size in [0, 1]. It is based on the chi-squared statistic
	but adjusted for the dimensions of the contingency table.

	## Theoretical Background

	Cramér's V measures the strength of association between two categorical variables
	using the chi-squared statistic from a contingency table:

	1. **Chi-squared statistic**: 
		χ² = Σ[(O_ij - E_ij)² / E_ij]
		where O_ij = observed counts, E_ij = expected counts under independence

	2. **Phi-squared**: 
		φ² = χ² / n
		where n = total number of observations

	3. **Cramér's V**:
		V = √[φ² / min(r - 1, k - 1)]
		where r = number of rows, k = number of columns in the contingency table

	The normalization by min(r-1, k-1) ensures V ranges from 0 to 1, making it
	comparable across different table sizes.

	## Interpretation

	- **V = 0**: No association (variables are independent)
	- **V = 1**: Perfect association
	- **0 < V < 1**: Increasing strength of association as V approaches 1

	## Effect Size Guidelines (Cohen, 1988)

	- **0.10**: Small association
	- **0.30**: Medium association  
	- **0.50**: Large association

	Parameters
	----------
	df : pd.DataFrame
		DataFrame containing the two categorical variables of interest.
	col_1 : str
		Name of the first categorical column.
	col_2 : str
		Name of the second categorical column.

	Returns
	-------
	float
		Cramér's V coefficient in [0, 1], or NaN if the table is degenerate
		(e.g. only one category in a variable).

	Raises
	------
	ValueError
		If there are fewer than 2 complete observations.

	"""
	data = df[[col_1, col_2]].dropna()

	if data.shape[0] < 2:
		raise ValueError("At least 2 complete observations are required.")

	# Observed contingency table
	cont_table = pd.crosstab(data[col_1], data[col_2])

	# Totals and expected counts
	observed = cont_table.to_numpy()
	n = observed.sum()
	row_totals = observed.sum(axis=1, keepdims=True)
	col_totals = observed.sum(axis=0, keepdims=True)

	expected = row_totals @ col_totals / n

	# Chi-squared statistic
	chi_2 = np.sum((observed - expected) ** 2 / expected)

	# Dimensions and normalisation
	r, k = cont_table.shape
	phi_squared = chi_2 / n
	min_dim = min(r - 1, k - 1)

	if min_dim <= 0:
		return float("nan")

	v = np.sqrt(phi_squared / min_dim)
	return float(v)
```

### Spearman Rank Correlation Between Binary and Ordinal Variables

The Spearman rank correlation coefficient (ρ) measures the **monotonic** relationship between two variables using ranks. When applied to a binary variable (coded as 0/1) and an ordinal variable, it computes the Pearson correlation between the binary variable and the ranks of the ordinal variable.

$$
\rho = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(r_i - \bar{r})}
{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2} \sqrt{\sum_{i=1}^{n} (r_i - \bar{r})^2}}
$$

Where:
- $x_i$ = binary variable values (0 or 1)
- $r_i$ = rank of the ordinal variable for observation $i$
- $\bar{x}$ = mean of the binary variable
- $\bar{r}$ = mean of the ranks
- $n$ = number of observations

## Interpretation
- **$\rho = +1$**: Perfect positive monotonic relationship (as binary increases from 0 to 1, ordinal ranks consistently increase)
- **$\rho = -1$**: Perfect negative monotonic relationship (as binary increases from 0 to 1, ordinal ranks consistently decrease)
- **$\rho = 0$**: No monotonic relationship between the binary indicator and ordinal variable
- **$0 < |\rho| < 0.3$**: Weak relationship
- **$0.3 \leq |\rho| < 0.7$**: Moderate relationship
- **$0.7 \leq |\rho| \leq 1$**: Strong relationship

## Key Characteristics for Binary vs Ordinal

**Spearman correlation uses rank ordering** rather than raw values:
- The binary variable is treated as numeric (0 or 1)
- The ordinal variable is converted to ranks (handling ties using average ranks)
- For categorical ordinal variables (e.g., 'Ex', 'Gd', 'TA'), values are sorted and mapped to ranks
- The coefficient measures monotonic association without assuming equal intervals

**Advantages over Pearson for Ordinal Data:**
- Does not assume equal spacing between ordinal categories
- More robust to outliers and non-linear monotonic relationships
- Only requires rank ordering, not interval-level measurement
- More appropriate when ordinal scale assumptions are questionable


```python
def spearman_binary_ordinal(
    df: pd.DataFrame, binary_col: str, ordinal_col: str
) -> float:
    """
    Calculate the Spearman rank correlation coefficient between a binary and an ordinal variable.

    The Spearman correlation measures the monotonic relationship between two variables using ranks.
    For a binary variable (coded as 0/1) and an ordinal variable, this computes the Pearson correlation
    between the binary variable and the ranks of the ordinal variable. This is equivalent to Spearman's rho
    when one variable is binary and the other is ordinal.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame containing the variables of interest.
    binary_col : str
        Name of the binary column (must contain only 0, 1, True, or False).
    ordinal_col : str
        Name of the ordinal column (can be numeric or categorical ordinal; will be ranked).

    Returns
    -------
    float
        Spearman correlation coefficient ranging from -1 to +1.

    Raises
    ------
    ValueError
        - If there are fewer than 3 complete observations.
        - If binary_col contains values other than 0, 1, True, or False.
        - If the ordinal variable has zero variance (all values identical).
        - If the binary variable has zero variance.

    Notes
    -----
    Interpretation:
    - +1: Perfect positive monotonic relationship.
    - -1: Perfect negative monotonic relationship.
    - 0: No monotonic relationship.

    Effect Size Guidelines (Cohen, 1988):
    - 0.10: Small effect
    - 0.30: Medium effect
    - 0.50: Large effect

    Mathematical Formula:
    Spearman rho = Pearson correlation between binary variable and ranks of ordinal variable.

    This implementation handles ordinal variables by ranking them (with ties handled by average ranks).
    For categorical ordinal variables (e.g., 'Ex', 'Gd', 'TA'), they are first converted to ranks based on
    their order in the data (assuming sorted order for simplicity; for precise ordinal mapping,
    preprocess the data externally).

    Examples
    --------
    >>> import pandas as pd
    >>> import numpy as np
    >>> data = pd.DataFrame({
    ...     'binary': [0, 1, 0, 1, 0],
    ...     'ordinal': [1, 3, 2, 4, 1]
    ... })
    >>> rho = spearman_binary_ordinal(data, 'binary', 'ordinal')
    >>> print(f"Spearman rho: {rho:.3f}")
    Spearman rho: 0.800
    """
    # 1. Subset & dropna
    data = df[[binary_col, ordinal_col]].dropna()
    if len(data) < 3:
        raise ValueError("At least 3 complete observations are required.")

    x = data[binary_col]
    y = data[ordinal_col]

    # 2. Validate that x is binary (0/1/True/False)
    unique_vals = set(pd.unique(x))
    if not unique_vals.issubset({0, 1, True, False}):
        raise ValueError(
            f"{binary_col!r} must contain only 0, 1, True, or False. "
            f"Found values: {unique_vals}"
        )

    # 3. Convert binary to numeric array
    x_num = x.astype(float).to_numpy()

    # 4. Handle ordinal variable: rank it
    # For categorical ordinals, sort unique values and map to ranks
    if y.dtype == "object":
        unique_ord = sorted(y.unique())
        rank_map = {val: rank for rank, val in enumerate(unique_ord, start=1)}
        y_ranks = y.map(rank_map).to_numpy()
    else:
        # Assume numeric ordinal; rank it
        y_ranks = pd.Series(y).rank(method="average").to_numpy()

    # 5. Check variances
    if np.var(x_num, ddof=1) <= 0:
        raise ValueError("Variance of binary_col is zero.")
    if np.var(y_ranks, ddof=1) <= 0:
        raise ValueError("Variance of ordinal_col ranks is zero.")

    # 6. Compute Pearson correlation between x_num and y_ranks
    x_mean = np.mean(x_num)
    y_mean = np.mean(y_ranks)
    numerator = np.sum((x_num - x_mean) * (y_ranks - y_mean))
    denominator = np.sqrt(np.sum((x_num - x_mean) ** 2)) * np.sqrt(
        np.sum((y_ranks - y_mean) ** 2)
    )
    if denominator == 0:
        return 0.0
    rho = numerator / denominator
    return float(rho)
```

### Spearman Correlation Formula

The Spearman correlation coefficient (ρ) measures the monotonic relationship between two variables. It is calculated as:

$$
\rho = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}
$$

Where:
- $d_i$ = difference between the ranks of corresponding variables
- $n$ = number of observations

## Interpretation
- **ρ = +1**: Perfect positive monotonic relationship (as one variable increases, the other always increases)
- **ρ = -1**: Perfect negative monotonic relationship (as one variable increases, the other always decreases)  
- **ρ = 0**: No monotonic relationship between variables

Unlike Pearson correlation which measures linear relationships, Spearman correlation captures any monotonic relationship (linear or nonlinear) by using rank orders rather than raw values, making it more robust to outliers.
