```python
from ucimlrepo import fetch_ucirepo

heart_disease = fetch_ucirepo(id=45)  # UCI Heart Disease dataset
heart_df = heart_disease.data.features
heart_target = heart_disease.data.targets
heart_metadata = heart_disease.metadata

heart_df.head()
```

```python
import types

import requests

# Fetch module code from URL
url = "https://raw.githubusercontent.com/jhlopesalves/data-science-notebooks/main/Python/projects/ames_house/data/missing_summary.py"
response = requests.get(url)  # Send HTTP GET request
response.raise_for_status()  # Raise exception for HTTP errors

module_code = response.text  # Extract response text

# Create new module object
mod = types.ModuleType("missing_summary")
# Execute module code in module's namespace
exec(module_code, mod.__dict__)

# Call function from imported module
mod.missing_data_summary(heart_df)
```
