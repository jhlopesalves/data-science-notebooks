```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/jhlopesalves/data-science-notebooks/refs/heads/main/Python/projects/book_recommender/data/books.csv"

books = pd.read_csv(url, usecols=lambda col: not col.startswith("Unnamed"))
```

```python
books
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)
ax.set(xlabel="Columns", ylabel="Missing Values")
plt.tight_layout()
plt.show()
```

```python
books["missing_description"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"] = 2025 - books["published_year"]
```

```python
columns_of_interest = [
	"num_pages",
	"age_of_book",
	"missing_description",
	"average_rating",
]

# Spearman correlation assesses how well the relationship between two variables can be described for both linear and non-linear relationships
correlation_matrix = books[columns_of_interest].corr(method="spearman")

fig, ax = plt.subplots(figsize=((10, 6)))
sns.heatmap(
	correlation_matrix,
	annot=True,
	fmt=".2f",
	cmap="Blues",
	cbar_kws={"label": "Spearman Correlation"},
)
ax.set_title("Correlation Heatmap")
plt.show()
```

```python
book_missing = books[
	~(books["description"].isna())
	& ~(books["num_pages"].isna())
	& ~(books["average_rating"].isna())
	& ~(books["published_year"].isna())
]
```

```python
book_missing["categories"].value_counts().reset_index().sort_values(
	by="count", ascending=False
)
```

```python
book_missing
```

```python
book_missing = book_missing.copy()
book_missing["words_in_description"] = book_missing["description"].str.split().str.len()
```

```python
fig, ax = plt.subplots(figsize=(12, 6))
sns.histplot(
	book_missing["words_in_description"],
	bins=50,
	color="C0",
	edgecolor="white",
	ax=ax,
)

ax.set(
	xlabel="Words in description",
	ylabel="Count",
	title="Distribution of words in book descriptions",
)
ax.grid(True, alpha=0.6, linewidth=0.4, color="grey")
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
book_missing_25_words = book_missing[book_missing["words_in_description"] >= 25].copy()
```

```python
book_missing_25_words
```

```python
book_missing_25_words["title_and_subtitle"] = np.where(
	book_missing_25_words["subtitle"].isna(),
	book_missing_25_words["title"],
	book_missing_25_words[["title", "subtitle"]].astype(str).agg(": ".join, axis=1),
)
```

```python
book_missing_25_words["title_and_subtitle"]
```

```python
book_missing_25_words["tagged_description"] = (
	book_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
)
```

```python
(
	book_missing_25_words.drop(
		["subtitle", "missing_description", "age_of_book", "words_in_description"],
		axis=1,
	).to_csv("data/books_cleaned.csv", index=False)
)
```
