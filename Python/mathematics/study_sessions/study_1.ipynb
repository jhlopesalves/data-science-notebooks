```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Part A — A removable discontinuity ("the hole")

We study

$$
g(x)=\frac{x^2-1}{x-1}.
$$

Step 1. Notice what happens at $x=1$.

* Numerator at $x=1$: $1^2-1=1-1=0$.
* Denominator at $x=1$: $1-1=0$.
* So at $x=1$ the expression becomes $0/0$, which is undefined. There is a "hole".

Step 2. Factor the numerator for all other $x$.

* $x^2-1=(x-1)(x+1)$.
* For any $x\neq 1$, we can cancel the $(x-1)$ factor with the denominator.
* So for $x\neq 1$,
	$$
	g(x)=x+1.
	$$

Step 3. What value does $x+1$ approach as $x$ gets close to 1 (from either side)?

* If $x=0.9$, $x+1=1.9$.
* If $x=0.99$, $x+1=1.99$.
* If $x=0.999$, $x+1=1.999$.
* If $x=1.001$, $x+1=2.001$.
* If $x=1.01$, $x+1=2.01$.
* If $x=1.1$, $x+1=2.1$.

Step 4. Read the pattern.
As $x$ gets closer to 1, $x+1$ gets closer to 2. From the left it looks like 1.9, 1.99, 1.999. From the right it looks like 2.001, 2.01, 2.1 moving down. Both sides head toward 2.

Conclusion.
$$
\lim_{x\to 1} g(x)=2,
$$
even though $g(1)$ is not defined. That is a removable discontinuity. The "hole" is at the point $(1,2)$.

## Part B — Limit exists, but the function value disagrees

Define a function
$$
f(x)=
\begin{cases}
x^2, & x\neq 2,\\
10, & x=2.
\end{cases}
$$

Step 1. Look near $x=2$, but not at $x=2$.
Compute $x^2$ for numbers near 2.

From the left

* $x=1.9$, $x^2=3.61$.
* $x=1.99$, $x^2=3.9601$.
* $x=1.999$, $x^2=3.996001$.

From the right

* $x=2.001$, $x^2=4.004001$.
* $x=2.01$, $x^2=4.0401$.
* $x=2.1$, $x^2=4.41$.

Step 2. Read the pattern.
Numbers a little below 2 give squares a little below 4.
Numbers a little above 2 give squares a little above 4.
So the values approach 4.

Step 3. Compare with the value at the point.
We defined $f(2)=10$. That is a single dot way above the nearby trend.

Conclusion.
$$
\lim_{x\to 2} f(x)=4,
$$
but $f(2)=10$. The limit exists and equals 4, yet the function value at the point is 10. That means a discontinuity. The graph would show a smooth parabola near $(2,4)$ with an odd black dot at $(2,10)$.

Key idea.
Limits describe the behaviour **near** the point. The actual value **at** the point can disagree and that does not change the limit.

## Part C — One-sided limits and a jump

Define the step function
$$
h(x)=
\begin{cases}
0, & x<0,\\
1, & x\ge 0.
\end{cases}
$$

Step 1. Approach 0 from the left ($x\to 0^-$).
When $x$ is negative, no matter how close to 0, $h(x)=0$.
Examples

* $x=-0.1\Rightarrow h(x)=0$.
* $x=-0.01\Rightarrow h(x)=0$.
* $x=-0.0001\Rightarrow h(x)=0$.

So the **left-hand limit** is
$$
\lim_{x\to 0^-} h(x)=0.
$$

Step 2. Approach 0 from the right ($x\to 0^+$).
When $x$ is positive or zero, $h(x)=1$.
Examples

* $x=0.0001\Rightarrow h(x)=1$.
* $x=0.01\Rightarrow h(x)=1$.
* $x=0.1\Rightarrow h(x)=1$.

So the **right-hand limit** is
$$
\lim_{x\to 0^+} h(x)=1.
$$

Step 3. Compare the sides.
Left-hand limit is 0. Right-hand limit is 1. They are not equal.

Conclusion.
The two-sided limit
$$
\lim_{x\to 0} h(x)
$$
does not exist. This is a **jump discontinuity**. The graph shows a horizontal line at height 0 for negative $x$, and a horizontal line at height 1 for non-negative $x$, with an open circle at $(0,0)$ and a filled dot at $(0,1)$.

## Part D — A gentle epsilon–delta demonstration with real numbers

Let's return to $g(x)=\frac{x^2-1}{x-1}$ for $x\neq 1$, which simplifies to $g(x)=x+1$ when $x\neq 1$. We already saw that
$$
\lim_{x\to 1} g(x)=2.
$$

The epsilon–delta idea puts numbers on "as close as you want".

Goal.
Show that we can make $g(x)$ fall inside any horizontal band around 2, no matter how tight, by picking $x$ close enough to 1.

Pick a specific tolerance.
Let's choose $\varepsilon=0.1$. This is a band from $2-0.1=1.9$ up to $2+0.1=2.1$.

We want to guarantee
$$
|g(x)-2|<0.1
$$
whenever $x$ is sufficiently close to 1 but not equal to 1.

Work it out step by step.

* Since $g(x)=x+1$ for $x\neq 1$,
	$$
	|g(x)-2|=|x+1-2|=|x-1|.
	$$
* So the condition $|g(x)-2|<0.1$ is the same as $|x-1|<0.1$.

Choose $\delta$.
If we choose $\delta=0.1$, then whenever $|x-1|<\delta$, we automatically get $|g(x)-2|<\varepsilon$.

Check with explicit numbers.

* Take $x=0.95$.
	Distance to 1 is $|0.95-1|=0.05<0.1$.
	Then $g(x)=x+1=1.95$. Distance to 2 is $|1.95-2|=0.05<0.1$. Good.
* Take $x=1.08$.
	Distance to 1 is $|1.08-1|=0.08<0.1$.
	Then $g(x)=2.08$. Distance to 2 is $|2.08-2|=0.08<0.1$. Good.
* Take $x=1.099$.
	Distance to 1 is $0.099<0.1$.
	$g(x)=2.099$. Distance to 2 is $0.099<0.1$. Good.

Tighten the band further.
Suppose you want $\varepsilon=0.001$. Then we need
$$
|g(x)-2|<0.001 \quad\Longleftrightarrow\quad |x-1|<0.001.
$$
Choose $\delta=0.001$.
Examples

* $x=0.9995\Rightarrow |x-1|=0.0005<0.001\Rightarrow g(x)=1.9995\Rightarrow |g(x)-2|=0.0005<0.001$.
* $x=1.0008\Rightarrow |x-1|=0.0008<0.001\Rightarrow g(x)=2.0008\Rightarrow |g(x)-2|=0.0008<0.001$.

Key insight.
For this function near this point, the output closeness equals the input closeness. That is why $\delta=\varepsilon$ works perfectly here. In other problems you might discover a different relationship and pick $\delta$ accordingly, but the spirit is the same. To force $g(x)$ within an epsilon band around 2, choose $x$ within a delta band around 1.


```python
# Build x-values approaching 1 from both sides, excluding 1 itself
x_left = 1 - np.array([1e-1, 1e-2, 1e-3, 1e-4])
x_right = 1 + np.array([1e-4, 1e-3, 1e-2, 1e-1])

# Compute g(x) = (x^2 - 1) / (x - 1) for x ≠ 1
g_left = (x_left**2 - 1) / (x_left - 1)
g_right = (x_right**2 - 1) / (x_right - 1)

# Table to see the numeric "approach"
df_g = pd.DataFrame({
	"x_left": x_left,
	"g(x_left)": g_left,
	"x_right": x_right,
	"g(x_right)": g_right,
})
df_g
```

```python
# Dense sampling around 1, excluding 1 exactly
x_dense = np.linspace(0.5, 1.5, 400)
x_dense = x_dense[x_dense != 1.0]
y_dense = (x_dense**2 - 1) / (x_dense - 1)

fig, ax = plt.subplots(figsize=(8, 5), dpi=140)
sns.lineplot(x=x_dense, y=y_dense, ax=ax)
ax.scatter(
	[1], [2], facecolors="none", edgecolors="black", s=120, linewidths=2
)  # open circle at the "hole"
ax.axvline(1, linestyle="--", linewidth=1)
ax.set_xlim(0.5, 1.5)
ax.set_ylim(1, 3)
ax.set_title(
	"Removable Discontinuity: g(x) = (x^2 - 1)/(x - 1), g undefined at x=1\nBut lim_{x→1} g(x) = 2"
)
ax.set_xlabel("x")
ax.set_ylabel("g(x)")
plt.show()
```

```python
# Curve y = x^2
x_curve = np.linspace(0, 3.5, 400)
y_curve = x_curve**2

fig, ax = plt.subplots(figsize=(8, 5), dpi=140)
sns.lineplot(x=x_curve, y=y_curve, ax=ax, linewidth=2)

# Open circle at the limiting value (2, 4) to show what nearby points suggest
ax.scatter([2], [4], facecolors="none", edgecolors="black", s=120, linewidths=2)

# Filled point at the actual (redefined) function value f(2) = 10
ax.scatter([2], [10], color="black", s=60)

ax.axvline(2, linestyle="--", linewidth=1)
ax.set_xlim(0, 3.5)
ax.set_ylim(0, 11)
ax.set_title("Limit vs. Function Value: lim_{x→2} x^2 = 4, but f(2) = 10")
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
plt.show()
```

```python
# A piecewise function with a jump at x = 0:
# h(x) = 0 for x < 0, and h(x) = 1 for x ≥ 0
x_left = -np.array([1e-1, 1e-2, 1e-3, 1e-4])
x_right = np.array([1e-4, 1e-3, 1e-2, 1e-1])

h_left = np.zeros_like(x_left)
h_right = np.ones_like(x_right)

pd.DataFrame({"x (← 0-)": x_left, "h(x)": h_left, "x (→ 0+)": x_right, "h(x)": h_right})
```

```python
x_neg = np.linspace(-1, 0, 200, endpoint=False)
x_pos = np.linspace(0, 1, 200)

y_neg = np.zeros_like(x_neg)
y_pos = np.ones_like(x_pos)

fig, ax = plt.subplots(figsize=(8, 5), dpi=140)
sns.lineplot(x=x_neg, y=y_neg, ax=ax)
sns.lineplot(x=x_pos, y=y_pos, ax=ax)

# Open and filled circles to mark the step at x=0
ax.scatter(
	[0], [0], facecolors="none", edgecolors="black", s=120, linewidths=2
)  # open at (0,0)
ax.scatter([0], [1], color="black", s=60)  # filled at (0,1)

ax.axvline(0, linestyle="--", linewidth=1)
ax.set_ylim(-0.5, 1.5)
ax.set_title(
	"One-Sided Limits at a Jump: left limit = 0, right limit = 1 → no two-sided limit"
)
ax.set_xlabel("x")
ax.set_ylabel("h(x)")
plt.show()
```

```python
# Same g(x) as before, dense sampling excluding x=1
x_dense = np.linspace(0.6, 1.4, 800)
x_dense = x_dense[x_dense != 1.0]
y_dense = (x_dense**2 - 1) / (x_dense - 1)

L = 2.0  # candidate limit
eps = 0.1  # height of the tolerance band

fig, ax = plt.subplots(figsize=(8, 5), dpi=140)
sns.lineplot(x=x_dense, y=y_dense, ax=ax)

# Horizontal epsilon-band: [L - eps, L + eps]
ax.axhspan(L - eps, L + eps, alpha=0.2)

# Visual markers at the centre
ax.axhline(L, linestyle="--", linewidth=1)
ax.axvline(1, linestyle="--", linewidth=1)

ax.set_xlim(0.6, 1.4)
ax.set_ylim(1.2, 2.8)
ax.set_title(
	"Epsilon Band: make g(x) lie within (L−ε, L+ε) by taking x close enough to 1"
)
ax.set_xlabel("x")
ax.set_ylabel("g(x)")
plt.show()
```
