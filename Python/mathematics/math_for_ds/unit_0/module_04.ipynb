### Unit 0: Mathematical Foundations & The Analytical Mindset
#### Module 0.4: The Idea of Limits

**(Phase 1: The Theoretical Lecture)**

The concept of a **limit** in mathematics allows us to describe the behavior of a function as its input gets arbitrarily close to a particular value, without necessarily reaching it. It's about approaching a destination, not necessarily arriving.

**1. The Intuitive Idea: Getting Closer and Closer**

Imagine you're walking towards a wall. You take steps, each covering half the remaining distance. 
* **Step 1:** You cover 1/2 the distance. 1/2 remains.
* **Step 2:** You cover half of the remaining 1/2, so 1/4 of the total. 1/4 remains.
* **Step 3:** You cover half of the remaining 1/4, so 1/8 of the total. 1/8 remains.
* ... and so on.

You get *infinitely close* to the wall, but theoretically, you never *quite* reach it because there's always a tiny fraction of the distance left to halve. The **limit** of your position, as the number of steps approaches infinity, is the wall itself.

**2. Formal Notation**

We write the limit of a function $f(x)$ as $x$ approaches a value $c$ as:

$$\lim_{x\to c} f(x) = L$$

This is read as "the limit of $f(x)$ as $x$ approaches $c$ equals $L$."

* $x \to c$: This means "$x$ gets closer and closer to $c$," from both sides (values slightly less than $c$ and values slightly greater than $c$), but $x$ does not necessarily have to equal $c$.
* $f(x)$: The function whose behavior we are observing.
* $L$: The value that $f(x)$ gets arbitrarily close to as $x$ approaches $c$.

**3. Why Not Just Plug In the Value? (The Need for Limits)**

For many simple, continuous functions, the limit *is* just the function's value at that point.
* **Example:** For $f(x) = x^2$, what is $\lim_{x\to 3} f(x)$?
    * As $x$ gets close to 3 (e.g., 2.9, 2.99, 3.01, 3.1), $f(x)$ gets close to $3^2 = 9$.
    * Here, $\lim_{x\to 3} x^2 = 9$, which is simply $f(3)$.

However, limits become essential in cases where plugging in the value directly is problematic:

* **Case 1: Holes in the Function (Removable Discontinuities)**
    Consider the function $f(x) = \frac{x^2 - 1}{x - 1}$.
    * What happens at $x=1$? Plugging in gives $\frac{1^2 - 1}{1 - 1} = \frac{0}{0}$, which is undefined.
    * But for any $x \neq 1$, we can simplify: $f(x) = \frac{(x-1)(x+1)}{x-1} = x+1$.
    * Now, let's look at the limit as $x$ *approaches* 1:
        * If $x = 0.9$, $f(x) = 1.9$.
        * If $x = 0.99$, $f(x) = 1.99$.
        * If $x = 1.1$, $f(x) = 2.1$.
        * If $x = 1.01$, $f(x) = 2.01$.
    * As $x$ gets arbitrarily close to 1 (but not equal to 1), $f(x)$ gets arbitrarily close to 2.
    * Therefore, $\lim_{x\to 1} \frac{x^2 - 1}{x - 1} = 2$.
    * The limit exists even though the function is undefined *at* the point. The graph looks like the line $y=x+1$ with a tiny hole at $(1, 2)$. 

* **Case 2: Jumps in the Function (Jump Discontinuities)**
    Consider a step function:
    $f(x) = \begin{cases} 0 & \text{if } x < 0 \\ 1 & \text{if } x \ge 0 \end{cases}$ 
    * What is the limit as $x$ approaches 0?
        * As $x$ approaches 0 from the left ($x = -0.1, -0.01, \dots$), $f(x)$ is always 0. The left-hand limit is 0. ($\lim_{x\to 0^-} f(x) = 0$)
        * As $x$ approaches 0 from the right ($x = 0.1, 0.01, \dots$), $f(x)$ is always 1. The right-hand limit is 1. ($\lim_{x\to 0^+} f(x) = 1$)
    * Since the limit from the left (0) does not equal the limit from the right (1), the overall limit $\lim_{x\to 0} f(x)$ **does not exist**.
    * A limit only exists if the function approaches the *same* value from both sides.

* **Case 3: Infinite Behavior (Asymptotes)**
    Consider $f(x) = \frac{1}{x^2}$.
    * What happens as $x$ approaches 0?
        * If $x = 0.1$, $f(x) = 1 / (0.01) = 100$.
        * If $x = 0.01$, $f(x) = 1 / (0.0001) = 10000$.
        * If $x = -0.1$, $f(x) = 1 / (0.01) = 100$.
    * As $x$ gets closer to 0 from either side, $f(x)$ grows infinitely large.
    * We say the limit is infinity: $\lim_{x\to 0} \frac{1}{x^2} = \infty$.
    * Technically, the limit doesn't exist as a finite number, but we use $\infty$ to describe the behavior. The graph has a vertical asymptote at $x=0$. 

**4. Limits at Infinity**

We can also ask what happens to a function as the input $x$ grows infinitely large (positive or negative).

**Notation:**
* $\lim_{x\to \infty} f(x) = L$ (as $x$ becomes arbitrarily large positive)
* $\lim_{x\to -\infty} f(x) = M$ (as $x$ becomes arbitrarily large negative)

* **Example:** $f(x) = \frac{1}{x}$.
    * As $x$ gets very large (100, 1000, 1,000,000), $f(x)$ gets closer and closer to 0 (0.01, 0.001, 0.000001).
    * $\lim_{x\to \infty} \frac{1}{x} = 0$.
    * Similarly, as $x$ gets very large negative (-100, -1000), $f(x)$ also gets closer to 0 (-0.01, -0.001).
    * $\lim_{x\to -\infty} \frac{1}{x} = 0$.
    * The graph has a horizontal asymptote at $y=0$. 

**5. Why Limits are Foundational for Calculus (and thus ML)**

The concept of a limit is the absolute cornerstone upon which calculus is built.

* **Derivatives (Rate of Change):** The derivative of a function at a point is defined as the *limit* of the slope of secant lines as the interval shrinks to zero. It formalizes the idea of "instantaneous rate of change." Gradient descent, the engine of training most ML models, relies entirely on calculating derivatives.
* **Integrals (Accumulation):** The definite integral is defined as the *limit* of the sum of areas of approximating rectangles as their width shrinks to zero. It formalizes the idea of "area under a curve." Integrals are fundamental in probability theory (calculating probabilities from density functions).
* **Continuity:** A function is formally defined as *continuous* at a point $c$ if $\lim_{x\to c} f(x) = f(c)$. That is, the limit exists, the function is defined at the point, and they are equal. Many theorems and algorithms assume continuity.

Understanding limits, even intuitively, is therefore essential for understanding *why* the tools of calculus (which underpin optimization and probability in ML) actually work.

**(Phase 2: The Code Approach Lecture)**

The core idea is to generate a sequence of numbers ($x$) getting closer and closer to our target value ($c$), evaluate the function $f(x)$ at each of these numbers, and see if the outputs approach a specific value ($L$).

**1. Generating Sequences Approaching a Point**

NumPy is excellent for this. We can use `np.linspace` to create points near $c$.

```python
# Sequence approaching 3 from the left (e.g., 2.9, 2.99, ...)
c = 3
num_points = 5
start = c - 1  # Start 1 unit away
x_left = np.linspace(start, c, num_points, endpoint=False)  # Exclude c itself
print(f"Approaching {c} from left: {x_left}")

# Better approach: Get exponentially closer
epsilon = 1e-6  # How close we ultimately want to get
x_left_better = c - np.logspace(0, -6, num_points)  # 10^0=1 down to 10^-6
print(f"Approaching {c} from left (better): {x_left_better}")

# Sequence approaching 3 from the right (e.g., 3.1, 3.01, ...)
x_right_better = (
	c + np.logspace(-6, 0, num_points)[::-1]
)  # 10^-6 up to 10^0, then reverse
print(f"Approaching {c} from right (better): {x_right_better}")
```

The `np.logspace` approach generates points that get *exponentially* closer, which is often better for visualizing limit behavior near a point.


**2. Evaluating the Function and Observing**

Let's test our simple case, $f(x) = x^2$, as $x \to 3$.

```python
def f(x):
	return x**2


c = 3
num_points = 7  # Increased for smoother approach

# Approaching from left
x_left = c - np.logspace(0, -6, num_points)  # 1 down to 1e-6
y_left = f(x_left)
print("\nApproaching 3 from left:")
for xi, yi in zip(x_left, y_left):
	print(f"  x = {xi:.7f}, f(x) = {yi:.7f}")

# Approaching from right
x_right = c + np.logspace(-6, 0, num_points)[::-1]  # 1e-6 up to 1, reversed
y_right = f(x_right)
print("\nApproaching 3 from right:")
for xi, yi in zip(x_right, y_right):
	print(f"  x = {xi:.7f}, f(x) = {yi:.7f}")
```

**Observation:** As $x$ gets closer to 3 from either side, $f(x)$ gets closer and closer to 9. This visually confirms our expectation that $\lim_{x\to 3} x^2 = 9$.

**3. Visualizing with Matplotlib**

Plots make the convergence much clearer.

```python
def f(x):
	return x**2


c = 3
num_points = 50  # More points for a smoother plot

x_left = c - np.logspace(0, -6, num_points)
y_left = f(x_left)

x_right = c + np.logspace(-6, 0, num_points)[::-1]
y_right = f(x_right)

# Create a figure and axis object for OOP approach
fig, ax = plt.subplots(figsize=(10, 6))

# Create a DataFrame for seaborn
df_left = pd.DataFrame({"x": x_left, "y": y_left, "direction": "Left"})
df_right = pd.DataFrame({"x": x_right, "y": y_right, "direction": "Right"})
df = pd.concat([df_left, df_right])

# Plot using seaborn with OOP style
sns.lineplot(
	data=df,
	x="x",
	y="y",
	hue="direction",
	ax=ax,
	palette={"Left": "blue", "Right": "red"},
	linewidth=2,
)

# Add limit lines
ax.axhline(9, color="black", linestyle="--", linewidth=1.5, label="Limit L=9")
ax.axvline(c, color="gray", linestyle=":", linewidth=1.5, label=f"x = {c}")

# Customize the plot
ax.set(title=f"Behavior of f(x) = x² as x approaches {c}", xlabel="x", ylabel="f(x)")
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, linewidth=0.6, color="grey")

# Improve aesthetics
sns.despine(ax=ax)
ax.tick_params(axis="both", which="major", labelsize=10)

plt.tight_layout()
plt.show()
```

The plot clearly shows both sequences of points converging towards the horizontal line $y=9$ as they get closer to the vertical line $x=3$.

**4. Investigating Problematic Cases**

  * **Case 1: Hole ($f(x) = (x^2 - 1) / (x - 1)$ as $x \to 1$)**

```python
def f_hole(x):
	# Handle potential division by zero for plotting near c
	# Note: For limit calculation, we conceptually avoid x=c
	numerator = x**2 - 1
	denominator = x - 1
	# Avoid division by zero warning using np.where
	return np.where(
		denominator != 0, numerator / denominator, np.nan
	)  # Return NaN at x=1


c = 1
limit_val = 2  # Our expected limit
num_points = 50

x_left = c - np.logspace(0, -6, num_points)
y_left = f_hole(x_left)

x_right = c + np.logspace(-6, 0, num_points)[::-1]
y_right = f_hole(x_right)

# Create a figure and axis object for OOP approach
fig, ax = plt.subplots(figsize=(10, 6))

# Create DataFrames for seaborn
df_left = pd.DataFrame({"x": x_left, "y": y_left, "direction": "Left"})
df_right = pd.DataFrame({"x": x_right, "y": y_right, "direction": "Right"})
df = pd.concat([df_left, df_right])

# Plot using seaborn with OOP style
sns.lineplot(
	data=df,
	x="x",
	y="y",
	hue="direction",
	ax=ax,
	palette={"Left": "blue", "Right": "red"},
	linewidth=2,
)

# Add limit lines and hole marker
ax.axhline(
	limit_val,
	color="black",
	linestyle="--",
	linewidth=1.5,
	label=f"Limit L={limit_val}",
)
ax.axvline(c, color="gray", linestyle=":", linewidth=1.5, label=f"x = {c}")
ax.plot(
	c,
	limit_val,
	"ko",
	markersize=8,
	fillstyle="none",
	label=f"Hole at ({c}, {limit_val})",
)

# Customize the plot
ax.set(
	title=r"Behavior of $f(x) = \frac{x^2-1}{x-1}$ as x approaches 1",
	xlabel="x",
	ylabel="f(x)",
)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, linewidth=0.6, color="grey")

# Improve aesthetics
sns.despine(ax=ax)
ax.tick_params(axis="both", which="major", labelsize=10)

plt.tight_layout()
plt.show()
```


The plot shows convergence towards $L=2$ from both sides, even though the function itself is undefined (plotted as a hole) at $x=1$.


  * **Case 2: Jump (Step Function)**

```python
def f_step(x):
	return np.where(x < 0, 0, 1)


c = 0
num_points = 50

x_left = c - np.logspace(-1, -6, num_points)  # Closer range for jump
y_left = f_step(x_left)

x_right = c + np.logspace(-6, -1, num_points)[::-1]
y_right = f_step(x_right)

# Create a figure and axis object for OOP approach
fig, ax = plt.subplots(figsize=(10, 6))

# Create DataFrames for seaborn
df_left = pd.DataFrame({"x": x_left, "y": y_left, "direction": "Left"})
df_right = pd.DataFrame({"x": x_right, "y": y_right, "direction": "Right"})
df = pd.concat([df_left, df_right])

# Plot using seaborn with OOP style
sns.lineplot(
	data=df,
	x="x",
	y="y",
	hue="direction",
	ax=ax,
	palette={"Left": "blue", "Right": "red"},
	linewidth=2,
)

# Add limit lines and jump visualization
ax.axhline(0, color="blue", linestyle="--", linewidth=1.5, label="Left limit L=0")
ax.axhline(1, color="red", linestyle="--", linewidth=1.5, label="Right limit L=1")
ax.axvline(c, color="gray", linestyle=":", linewidth=1.5, label=f"x = {c}")

# Customize the plot
ax.set(title="Step Function Behavior near x = 0", xlabel="x", ylabel="f(x)")
ax.set_ylim(-0.5, 1.5)  # Adjust y-limits for clarity
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, linewidth=0.6, color="grey")

# Improve aesthetics
sns.despine(ax=ax)
ax.tick_params(axis="both", which="major", labelsize=10)

plt.tight_layout()
plt.show()
```

The plot clearly shows the function approaching different values (0 from left, 1 from right), illustrating why the limit does not exist at $x=0$.



**5. Limits at Infinity**

We can simulate approaching infinity by using large numbers.

```python
def f_inv(x):
	return 1 / x


# Sequence approaching +infinity
x_pos_inf = np.logspace(1, 6, 100)  # 10^1 to 10^6
y_pos_inf = f_inv(x_pos_inf)

# Sequence approaching -infinity
x_neg_inf = -np.logspace(1, 6, 100)[::-1]  # -(10^6) to -(10^1)
y_neg_inf = f_inv(x_neg_inf)

# Create DataFrames for seaborn
df_pos = pd.DataFrame({"x": x_pos_inf, "y": y_pos_inf, "direction": r"$x \to \infty$"})
df_neg = pd.DataFrame({"x": x_neg_inf, "y": y_neg_inf, "direction": r"$x \to -\infty$"})
df = pd.concat([df_pos, df_neg])

# Create a figure and axis object for OOP approach
fig, ax = plt.subplots(figsize=(10, 6))

# Plot using seaborn with OOP style
sns.lineplot(
	data=df,
	x="x",
	y="y",
	hue="direction",
	ax=ax,
	palette={r"$x \to \infty$": "red", r"$x \to -\infty$": "blue"},
	linewidth=2,
)

# Add limit line
ax.axhline(0, color="black", linestyle="--", linewidth=1.5, label="Limit L=0")

# Customize the plot
ax.set(
	title=r"Behavior of $f(x) = \frac{1}{x}$ as $x \to \pm \infty$",
	xlabel="x",
	ylabel="f(x)",
)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, linewidth=0.6, color="grey")

# Improve aesthetics
sns.despine(ax=ax)
ax.tick_params(axis="both", which="major", labelsize=10)

plt.tight_layout()
plt.show()
```


The plot shows the function values getting closer and closer to 0 as $x$ moves further away from the origin in either the positive or negative direction, confirming $\lim_{x\to \pm \infty} \frac{1}{x} = 0$.

This computational approach doesn't *prove* limits, but it provides strong visual and numerical intuition for the concept of approaching a value. It helps visualize why limits are necessary for dealing with points where functions might be undefined or behave strangely, and how we describe long-term trends using limits at infinity.
